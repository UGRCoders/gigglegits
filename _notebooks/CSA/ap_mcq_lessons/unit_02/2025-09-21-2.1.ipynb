Unit 1 built algorithms out of statements that just ran one after another. This unit introduces two new ways for a program to decide what runs: choosing between paths, and repeating a path more than once.

## Three Building Blocks

Every algorithm, no matter how complicated, is built out of exactly three things:

| Building block | What it means |
|---|---|
| Sequence | Statements run one after another, in the order they're written |
| Selection | The program picks one path or another based on a condition |
| Repetition | A block of code runs more than once |

Everything you wrote in Unit 1, variable declarations, expressions, method calls, was sequence. This unit is about the other two.

## A Preview of Selection

Selection uses an `if` statement to run code only when a condition is true. You'll learn the full syntax in the next couple of lessons, but here's what it looks like:

```java
int temperature = 45;

if (temperature < 32) {
    System.out.println("It's freezing outside.");
} else {
    System.out.println("It's above freezing.");
}
```

Only one of those two `System.out.println` lines actually runs. Which one depends entirely on the value of `temperature` when the program executes, not on which line appears first in the code.

## A Preview of Repetition

Repetition uses a loop, like `while` or `for`, to run a block of code multiple times without copy pasting it over and over:

```java
for (int i = 1; i <= 5; i++) {
    System.out.println("Lap " + i);
}
```

This prints five lines, `Lap 1` through `Lap 5`, from a single `System.out.println` statement written once. Without repetition, printing five laps would mean writing that line five separate times.

### The Big Shift From Unit 1

In a purely sequential program, you can predict exactly what runs just by reading top to bottom. Once selection and repetition are involved, the order code executes in can differ from the order it's written in, and the same line can run zero times, once, or many times depending on the data. Learning to trace through code, rather than just read it top to bottom, is the main new skill this unit builds.

## What's Coming in This Unit

Selection and repetition each need their own building blocks before you can combine them into real algorithms. This unit works through them in order: boolean expressions, then `if` statements (with nesting and compound conditions), then loops (`while` and `for`), and finally putting it all together to implement real selection and iteration algorithms, including ones that work over strings and nested loops.

The next lesson starts at the foundation both selection and repetition are built on: boolean expressions, the true or false values that every condition ultimately comes down to.
