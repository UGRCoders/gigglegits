### Question 6

```java
for (int i = 0; i < 5; i++) {
    for (int j = 0; j < 3; j++) {
        System.out.println("*"); // statement X
    }
}
```

How many times does statement X execute?<br>
a) 5<br>
b) 3<br>
c) 8<br>
d) 15

<details>
<summary>Show answer</summary>

d) 15. The outer loop runs 5 times, and for each of those, the inner loop runs 3 times, so 5 x 3 = 15.

</details>

````python
// Q6 Hack: Add a counter variable that increments once per execution of statement X,
// then print the counter after both loops finish to confirm the total.
int count = 0;
for (int i = 0; i < 5; i++) {
    for (int j = 0; j < 3; j++) {
        count++;
    }
}
System.out.println(count);
````

### Question 7

```java
String word = "planet";
String sub = "net";

for (int i = 0; i < word.length(); i++) {
    String portion = word.substring(i, i + sub.length());
    if (portion.equals(sub)) {
        System.out.println("found");
        break;
    }
}
```

What is wrong with this loop, even though it happens to find "net" before the bug matters?<br>
a) `substring` cannot be used in a loop<br>
b) The loop bound `i < word.length()` lets `i` get large enough that `word.substring(i, i + sub.length())` throws an exception on later iterations<br>
c) `.equals()` should be `==` for Strings<br>
d) Nothing is wrong with this loop

<details>
<summary>Show answer</summary>

b) The loop bound should be `i <= word.length() - sub.length()`. As written, once `i + sub.length()` exceeds `word.length()`, `substring` throws a `StringIndexOutOfBoundsException`. This particular example happens to `break` out early because it finds a match first, but the loop is still unsafe in general.

</details>

````python
// Q7 Hack: Change sub to a string that is NOT found anywhere in word (like "xyz")
// and run the loop to see the StringIndexOutOfBoundsException. Then fix the loop bound
// to i <= word.length() - sub.length() and confirm it no longer crashes.
String word = "planet";
String sub = "xyz";

for (int i = 0; i < word.length(); i++) {
    String portion = word.substring(i, i + sub.length());
    if (portion.equals(sub)) {
        System.out.println("found");
        break;
    }
}
System.out.println("not found");
````
