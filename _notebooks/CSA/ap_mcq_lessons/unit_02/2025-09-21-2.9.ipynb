# Implementing Selection and Iteration Algorithms

You now have both halves of this unit's toolkit: if statements and loops. This lesson combines them into a few standard algorithm patterns that show up constantly: finding a maximum, counting matches, and summing values that meet a condition.

<div class="lesson-wrap">

These examples loop over an `ArrayList`, using just `.size()` and `.get(index)`. `ArrayList` gets its own full lesson in Unit 4, but a for loop that walks through one works exactly like the for loops you already know: initialize an index at 0, keep going while it's less than `.size()`, and increment.

## Standard Algorithm: Finding the Maximum

```python
import java.util.ArrayList;

public class FindMax {
    public static void main(String[] args) {
        ArrayList<Integer> scores = new ArrayList<Integer>();
        scores.add(72);
        scores.add(95);
        scores.add(88);
        scores.add(60);

        int max = scores.get(0); // start with the first value, not 0

        for (int i = 1; i < scores.size(); i++) {
            if (scores.get(i) > max) {
                max = scores.get(i);
            }
        }

        System.out.println("Max: " + max);
    }
}
FindMax.main(null);
```

<div class="callout callout-hint">
<h4>Common Mistake: Starting max at 0</h4>

Starting `max` at `0` seems harmless until every value in the list is negative, in which case `0` is bigger than all of them and the algorithm reports a wrong maximum that was never even in the list. Always start `max` (or `min`) at the first actual element, then loop starting from index `1` so you don't compare it against itself.

</div>

## Standard Algorithm: Counting Matches

```python
import java.util.ArrayList;

public class CountPassing {
    public static void main(String[] args) {
        ArrayList<Integer> scores = new ArrayList<Integer>();
        scores.add(72);
        scores.add(95);
        scores.add(58);
        scores.add(60);
        scores.add(88);

        int count = 0;
        for (int i = 0; i < scores.size(); i++) {
            if (scores.get(i) >= 60) {
                count++;
            }
        }

        System.out.println("Passing scores: " + count);
    }
}
CountPassing.main(null);
```

The pattern is: a counter variable starting at 0, a loop over every element, and an if statement that increments the counter only when the condition is met.

## Standard Algorithm: Conditional Sum

Summing works the same way, just adding the value itself instead of adding 1 to a counter:

```python
import java.util.ArrayList;

public class SumPassing {
    public static void main(String[] args) {
        ArrayList<Integer> scores = new ArrayList<Integer>();
        scores.add(72);
        scores.add(95);
        scores.add(58);
        scores.add(60);
        scores.add(88);

        int total = 0;
        for (int i = 0; i < scores.size(); i++) {
            if (scores.get(i) >= 60) {
                total += scores.get(i);
            }
        }

        System.out.println("Total of passing scores: " + total);
    }
}
SumPassing.main(null);
```

All three algorithms share the same skeleton: set up a starting value before the loop, walk through every element once, and use an if statement inside the loop to decide whether that element affects the running result.
