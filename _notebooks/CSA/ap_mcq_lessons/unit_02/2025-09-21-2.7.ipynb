# While Loops

A while loop repeats a block of code for as long as its condition stays true, checking the condition again before every repetition, including the first.

## How a While Loop Works

A while loop has one part: a boolean condition. Before every pass through the loop body, Java checks that condition. If it is true, the body runs and the loop checks again. If it is false, the loop ends and execution moves to the line after it. Run the cell below:

Three things have to be true for this to behave correctly: a variable is set up before the loop (`count = 0`), the condition depends on that variable (`count < 5`), and something inside the loop body changes that variable (`count++`). Miss the last part and you get an infinite loop, since the condition never becomes false.

````java
int count = 0;

while (count < 5) {
    System.out.println(count);
    count++;
}
````

## Sentinel-Controlled Loops

A while loop is the right choice when you do not know in advance how many times the loop needs to run. A common pattern is a sentinel-controlled loop, where the loop keeps going until it sees a specific "stop" value.

```java
import java.util.Scanner;

Scanner scanner = new Scanner(System.in);
int total = 0;
int next = scanner.nextInt();

while (next != -1) {
    total += next;
    next = scanner.nextInt();
}

System.out.println("Total: " + total);
```

Here `-1` is the sentinel value. The loop has no idea ahead of time how many numbers the user will type, only that it should stop once it sees `-1`. A for loop would be awkward here, since a for loop header expects to know its stopping point up front, which is exactly what this problem does not have.

## A More Involved Example

While loops can be nested inside each other just like if statements can. This example builds a small pyramid of asterisks, using one while loop to print the spaces before each row's stars and another to print the stars themselves.

````java
public class PyramidPattern {
    public static void main(String[] args) {
        int height = 5;
        int row = 1;

        while (row <= height) {
            int spaces = height - row;
            int stars = 2 * row - 1;

            // Print spaces
            int spaceCount = spaces;
            while (spaceCount > 0) {
                System.out.print(" ");
                spaceCount--;
            }

            // Print stars
            int starCount = stars;
            while (starCount > 0) {
                System.out.print("*");
                starCount--;
            }

            System.out.println(); // Move to the next line for the next row
            row++;
        }
    }
}
PyramidPattern.main(null);
````

<h4>Common Mistake: the Infinite Loop</h4>

The most common while loop bug is forgetting to update the variable the condition depends on. This never stops:

```java
int count = 0;
while (count < 5) {
    System.out.println(count);
    // forgot count++
}
```

Every check of `count < 5` sees the same value, so the loop never becomes false. Before you run a while loop, always trace through by hand and confirm the condition variable actually changes inside the body, and that it changes toward making the condition false, not away from it.

</div>

```java
int n = 20;
while (n > 1) {

}
```

It prints `20`, `10`, `5`, `2`, in that order, then stops, since `2 / 2 = 1` and `1 > 1` is false. The body runs 4 times.
