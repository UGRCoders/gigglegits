## && and ||

| Operator | Name | True when |
|---|---|---|
| `&&` | and | both sides are true |
| `\|\|` | or | at least one side is true |
| `!` | not | flips true to false and false to true |

Here's the nested if from the last lesson, rewritten as one if statement using `&&`:

```java
int age = 17;
boolean hasPermit = true;

// nested version (last lesson)
if (age >= 16) {
    if (hasPermit) {
        System.out.println("You can drive with a licensed adult.");
    }
}

// same logic, one condition
if (age >= 16 && hasPermit) {
    System.out.println("You can drive with a licensed adult.");
}
```

Both versions only print when `age >= 16` and `hasPermit` are both true. `&&` requires every part to be true; `||` only needs one:

```java
int day = 6; // 1 = Monday ... 7 = Sunday

if (day == 6 || day == 7) {
    System.out.println("It's the weekend.");
}
```

## The ! Operator

`!` negates a single boolean, flipping it to its opposite:

```java
boolean isRaining = false;

if (!isRaining) {
    System.out.println("Good day for a walk.");
}
```

`!isRaining` is `true` exactly when `isRaining` is `false`.

## Short-Circuit Evaluation

Java evaluates `&&` and `||` left to right, and stops early the moment the overall result is already determined. For `&&`, if the left side is `false`, the whole expression must be `false`, so Java never even checks the right side. For `||`, if the left side is `true`, the whole expression must be `true`, and the right side is skipped.

This isn't just an efficiency trick, it changes which order is safe to write conditions in:

```java
String name = null;

// safe: name != null is checked first
if (name != null && name.length() > 0) {
    System.out.println("Name: " + name);
}

// unsafe: name.length() runs before name is confirmed non-null
if (name.length() > 0 && name != null) {
    System.out.println("Name: " + name);
}
```

<div class="callout callout-hint">
<h4>Common Mistake: Wrong Operand Order</h4>

The second version above throws a `NullPointerException` before Java ever gets to check `name != null`, since `name.length()` runs first and `name` is `null`. Whenever one condition needs to be true for a second condition to be safe to check, put the guard condition first.

</div>
