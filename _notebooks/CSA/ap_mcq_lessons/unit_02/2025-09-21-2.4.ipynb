# Nested if Statements

A nested if statement is an if statement placed inside the body of another if (or else) statement, letting a program test one condition only after another has already been confirmed true. An else if chain, which you may have already seen, is actually a special case of this same idea.

## Nesting an if Inside Another if

Once you know an outer condition is true, you can put a whole second if statement inside that block to test something else, but only for the cases where the outer condition already held.

```java
int age = 17;
boolean hasPermit = true;

if (age >= 16) {
    if (hasPermit) {
        System.out.println("You can drive with a licensed adult.");
    } else {
        System.out.println("You need a permit first.");
    }
} else {
    System.out.println("You are too young to drive.");
}
```

The inner if only runs at all when `age >= 16` is true. If `age` is 15, the program never even looks at `hasPermit`. Later in this unit you will learn compound boolean expressions (`&&`, `||`), which let you combine both conditions into a single if statement. Nesting is still useful when the inner check only makes sense inside a specific branch, or when each branch needs its own separate follow up logic.

## Else if Chains Are Nested ifs in Disguise

An else if chain looks like a flat list of options, but it is really an if nested inside the else of the previous one. These two snippets behave identically:

```java
// else if chain
if (score >= 90) {
    grade = "A";
} else if (score >= 80) {
    grade = "B";
} else {
    grade = "C";
}
```

```java
// the same logic, written as an explicitly nested if
if (score >= 90) {
    grade = "A";
} else {
    if (score >= 80) {
        grade = "B";
    } else {
        grade = "C";
    }
}
```

Java lets you drop the braces around that inner `if` inside the `else`, which is exactly what turns the second version into the more familiar `else if` syntax.

### Flow of Execution

Each condition in an else if chain is evaluated in the order it is written. The first true condition's code runs, and the rest are skipped, since they are nested inside the else branches that never got reached.

### Key Concept

The order of conditions matters. More specific conditions should come before broader ones to ensure accurate results.

## Common Mistake: the Dangling Else

When you nest an if without braces, Java always attaches an `else` to the closest unmatched `if`, not the one it looks visually lined up with. This trips people up constantly:

```java
int x = 5;

if (x > 0)
    if (x > 10)
        System.out.println("big positive");
    else
        System.out.println("this else belongs to \"x > 10\", not \"x > 0\"");
```

Even though the indentation makes it look like the `else` matches `if (x > 0)`, it actually matches `if (x > 10)`, since that is the nearest if without its own else. With `x = 5`, nothing prints at all. Always use braces around nested if bodies so the pairing is explicit and matches what the indentation shows.
