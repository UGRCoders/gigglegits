# Comparing Boolean Expressions

Two boolean expressions can look completely different and still always produce the same result. This lesson covers how to prove that with a truth table, and a pair of rules, De Morgan's Laws, for negating compound expressions correctly.

<h4>Objectives</h4>

- State De Morgan's Laws for negating `&&` and `||` expressions
- Correctly negate a compound boolean expression

## Truth Tables

A truth table lists every possible combination of `true`/`false` for each variable in an expression, and shows what the expression evaluates to for each combination. With two variables, `p` and `q`, there are four combinations:

| p | q | p && q | p \|\| q |
|---|---|---|---|
| true | true | true | true |
| true | false | false | true |
| false | true | false | true |
| false | false | false | false |

Two expressions are equivalent if their columns match on every row, for every combination of inputs, not just the one you happen to test.

## De Morgan's Laws

De Morgan's Laws describe exactly what happens when you negate a compound expression:

- `!(p && q)` is equivalent to `!p || !q`
- `!(p || q)` is equivalent to `!p && !q`

In words: negating an "and" flips it to an "or" of the negated parts, and negating an "or" flips it to an "and" of the negated parts. Run the cell below and confirm each pair of lines matches:

````python
boolean p = true;
boolean q = false;

System.out.println(!(p && q));   // true
System.out.println(!p || !q);    // true, matches !(p && q)

System.out.println(!(p || q));   // false
System.out.println(!p && !q);    // false, matches !(p || q)
````

## Applying It: Rewriting a Guard Condition

Suppose code only proceeds when a user is both logged in and has a verified email:

```java
if (isLoggedIn && isVerified) {
    System.out.println("Access granted.");
}
```

To write the opposite condition, the one where access should be denied, apply De Morgan's Law rather than guessing:

```java
if (!isLoggedIn || !isVerified) {
    System.out.println("Access denied.");
}
```

<div class="callout callout-hint">
<h4>Common Mistake: Distributing ! Without Flipping the Operator</h4>

It's tempting to negate `isLoggedIn && isVerified` by just slapping a `!` on each piece and keeping `&&`: `!isLoggedIn && !isVerified`. That is not the same expression. It's only true when both are false, when really you want it true whenever either one fails. The operator has to flip along with each variable, `&&` becomes `||`, exactly as De Morgan's Laws describe.
