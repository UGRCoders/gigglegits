## Statement Execution Counts

A statement execution count is the number of times a particular line of code actually runs while the program executes. A statement outside any loop has an execution count of 1. A statement inside a loop that runs 10 times has an execution count of 10, unless it's further guarded by an if, in which case you only count the passes where the condition is also true.

Execution counts give you a quick, informal way to compare two pieces of code without timing them: the version with fewer total executions of its slowest statement is generally the faster one.

````python
// CODE EXAMPLE #1 (for loop)
public class InformalCodeAnalysis {
    public static void main(String[] args) {
        int count = 0;
        for (int k = 0; k < 30; k++)
        {
            if (k % 3 == 0) // statement 1
            {
                count++; // statement 2
            }
        }
    }
}
````

<div class="callout callout-practice">
<h4>Check Your Understanding</h4>

How many times will statement 1 execute?

How many times will statement 2 execute?

<details>
<summary>Show answer</summary>

Statement 1 (`k % 3 == 0`) is checked on every pass through the loop, so it executes 30 times, once for each value of `k` from 0 to 29. Statement 2 (`count++`) only runs when that check is true, which happens for `k` = 0, 3, 6, 9, ..., 27, so it executes 10 times.

</details>

</div>

````python
// CODE EXAMPLE #2 (for loop)
// Rewrite the code segment below to have a faster run-time based on statement execution counts
for (int k = 0; k < 135; k++)
{
    if (k % 5 == 0)
    {
        System.out.println(k);
    }
}
````

<details>
<summary>Show sample answer</summary>

The loop condition `k < 135` forces the loop to check 135 times, even though only every fifth value actually gets printed (`k` = 0, 5, 10, ..., 130, which is 27 executions of the print statement). Stepping the loop variable by 5 directly avoids the wasted checks:

```java
for (int k = 0; k < 135; k += 5) {
    System.out.println(k);
}
```

This version's loop check now executes only 27 times instead of 135, since it never visits the values that would have failed the `k % 5 == 0` test anyway.

</details>

````python
// CODE EXAMPLE #3 (while loop)

int num = (int)(Math.random() * 10);
while (num % 2 != 0)
{
    num = (int)(Math.random() * 10); // statement 4
}
````

<div class="callout callout-practice">
<h4>Check Your Understanding</h4>

What is the min and max number of times statement 4 will execute?

<details>
<summary>Show answer</summary>

Minimum: 0. If the very first random value assigned to `num` before the loop happens to already be even, the loop condition is false immediately and the body, including statement 4, never runs.

Maximum: there is no fixed maximum. Since `num` is reassigned to a new random value each pass, it is always possible (just extremely unlikely) to keep rolling odd numbers indefinitely. This is a good example of a loop whose execution count cannot be pinned to an exact number in advance, only bounded on one side.

</details>

</div>

````python
// CODE EXAMPLE #4 (nested loop)

for (int outer = 0; outer < 3; outer++)
{
    for (int inner = 0; inner < 4; inner++)
    {
        // statement #5
    }
}
````

<div class="callout callout-practice">
<h4>Check Your Understanding</h4>

How many times will statement 5 execute?

<details>
<summary>Show answer</summary>

12 times. The outer loop runs 3 times, and for each of those, the inner loop runs 4 times, so the total is 3 x 4 = 12, using the same nested loop multiplication rule from the previous lesson.

</details>

Execution counts are the informal groundwork for run-time analysis you'll see described more formally, with terms like Big-O, in later courses. For now, being able to look at a loop or a nested loop and say roughly how many times it runs is enough to compare two approaches and pick the faster one.
