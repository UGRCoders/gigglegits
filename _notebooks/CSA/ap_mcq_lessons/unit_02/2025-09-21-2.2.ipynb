## What Is a Boolean Expression?

A boolean expression is any expression that evaluates to exactly one of two values: `true` or `false`. In Java, the type for those values is `boolean`.

```java
boolean isPassing = true;
boolean hasHomework = false;

System.out.println(isPassing);
System.out.println(hasHomework);
```

That's a boolean literal assigned directly. More often, a boolean expression comes from comparing two values.

## Relational Operators

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `<` | less than |
| `>` | greater than |
| `<=` | less than or equal to |
| `>=` | greater than or equal to |

```java
int score = 87;

System.out.println(score == 87);   // true
System.out.println(score != 87);   // false
System.out.println(score > 90);    // false
System.out.println(score >= 87);   // true
```

Each of these lines prints a single `boolean` value. That value is exactly what an `if` statement or a loop condition checks.

<div class="callout callout-hint">
<h4>Common Mistake: = vs ==</h4>

A single `=` assigns a value. A double `==` compares two values. Mixing them up is one of the most common bugs in beginner Java:

```java
int score = 87;

if (score = 90) { // does not compile: = is assignment, not comparison
    System.out.println("Perfect score!");
}
```

Java actually catches this particular case for you at compile time, since `score = 90` evaluates to an `int`, not a `boolean`, and an `if` requires a `boolean`. Get in the habit of reading `==` out loud as "is equal to" and `=` as "gets set to" so the two never blur together.

</div>

## Comparing Strings

`==` works correctly for primitives like `int`, `double`, and `char`, but not for comparing the contents of two `String` objects. Recall from Unit 1 that a `String` is a reference type: a `String` variable holds a reference to an object on the heap, not the characters themselves. `==` checks whether two references point to the exact same object, not whether the text inside them matches.

```java
String a = "hello";
String b = "hello";
String c = new String("hello");

System.out.println(a == b); // true, both refer to the same literal
System.out.println(a == c); // false, c is a separate object in memory
System.out.println(a.equals(c)); // true, compares the actual characters
```

Use `.equals()` whenever you're comparing what's inside two Strings, and save `==` for primitives and for checking whether two references point to the same object on purpose.
