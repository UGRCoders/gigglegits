# For Loops

A for loop repeats a block of code a known number of times, using a single header that sets it up, checks it, and updates it.

## For Loops vs. While Loops

- A for loop is an iterative statement that checks for a condition and repeatedly executes a block of code as long as the condition is met.
- The condition specifies how many times the loop runs.
- While loops: use when the number of iterations is unknown.
- For loops: use when the number of iterations is known.

## The Same Loop, Two Ways

Any for loop can be rewritten as a while loop, and vice versa. These two loops print the same thing:

```python
int i = 0;
while (i < 5) {
    System.out.println(i);
    i++;
}
```

```python
for (int i = 0; i < 5; i++) {
    System.out.println(i);
}
```

The for loop version packs the same three pieces (initialize `i`, check `i < 5`, update `i`) into one header instead of spreading them across three lines. Since we already know exactly five iterations are needed, the for loop is the more natural choice here.

## The Three Parts of a For Loop Header

A for loop header has three parts: variable initialization, a Boolean (conditional) expression, and an increment or decrement statement.

<div class="callout callout-practice">
<h4>Which Part Is Which?</h4>

- Variable initialization (`int i = 0`): sets the variable before the loop starts.
- Boolean (conditional) expression (`i < 5`): defines the condition for the loop to run. In this case, the loop continues as long as `i` is less than 5, so it loops 5 times, for `i` equal to 0 through 4.
- Increment or decrement statement (`i++`): changes the variable each time the code block in the loop finishes executing. In this case it increases `i` by 1.

</div>

The loop variable can also be used inside the code block for other purposes besides controlling how many times the loop repeats. Together, the Boolean expression and the increment or decrement statement determine how many times the loop will repeat.

## Combining a For Loop With an if Statement

A for loop's body is just a block of code, so it can hold an if statement like any other block. Watch what happens when the if body doesn't use braces:

```python

int sum = 0;
int sum_all = 0;

for (int i = 0; i < 10; i++) {
    if (i % 2 == 0)
        sum += i;
    sum_all += i;
}

System.out.println(sum);
System.out.println(sum_all);
```

Without braces, the if statement only controls the very next line, `sum += i`. The line `sum_all += i` is not part of the if body at all, it just happens to be indented to look like it is, so it runs on every pass through the loop regardless of the condition. `sum` ends up as 20 (the even numbers 0, 2, 4, 6, 8), while `sum_all` ends up as 45 (every number 0 through 9). This is the same brace scoping rule from the nested if lesson, and it applies just as easily to loop bodies.
