## What Is Nested Iteration?

Nested iteration occurs when we have a loop inside of another loop, similar to nested conditional statements.

## How Nested Loops Execute

When you have one loop inside another, the inner loop has to finish all its rounds before the outer loop moves to the next round. If the inner loop has a "stop" command, it only stops for that round of the outer loop. The next time the outer loop starts a new round, the inner loop starts over.

If you have two nested loops without stops, and the outer one runs n times while the inner one runs m times each time the outer one goes around, the inner loop will run m times n times, which is m times n in total. This rule also applies if you have more than two nested loops. To find the total number of times the innermost loop runs, just multiply how many times each loop runs per round.

````python
public class NestedLoopsDemo {
    public static void main(String[] args) {
        int n = 3; //numb of times the outside loop runs
        int m = 2; //numb of times the inside loop runs

        //the nested loops
        for (int i = 1; i <= n; i++) {
            System.out.println("Outer loop iteration: " + i);
            for (int j = 1; j <= m; j++) {
                System.out.println("Inner loop iteration: " + j);
            }
        }
    }
}
NestedLoopsDemo.main(null)
````

Trace through it by hand: the outer loop starts at `i = 1`. Before `i` ever becomes `2`, the entire inner loop runs to completion, printing `j = 1` and `j = 2`. Only then does control return to the outer loop, `i` becomes `2`, and the inner loop restarts from `j = 1` again. That "restart from the beginning every time" behavior is what makes nested loops multiply instead of add.

## Common Mistake: Reusing the Same Loop Variable

Each nested loop needs its own loop variable. Trying to reuse the same name for both fails to compile, since the inner loop's declaration would collide with the outer one that is still in scope:

```java
// does not compile
for (int i = 0; i < 3; i++) {
    for (int i = 0; i < 4; i++) { // error: i is already defined
        System.out.println(i);
    }
}
```

The fix is simply to give each loop a distinct variable, conventionally `i` for the outer loop and `j` for the next one in, `k` after that, and so on.

## Building a Multiplication Table

Nested loops are the natural way to generate anything laid out in rows and columns. Here, the outer loop picks a row and the inner loop fills in every column of that row:

```java
public class MultiplicationTable {
    public static void main(String[] args) {
        for (int row = 1; row <= 5; row++) {
            for (int col = 1; col <= 5; col++) {
                System.out.print(row * col + "\t");
            }
            System.out.println();
        }
    }
}
MultiplicationTable.main(null);
```

Notice the `System.out.println()` with nothing in it sits outside the inner loop but inside the outer one. It runs once per row, after that row's columns are all filled in, which is what moves output to a new line between rows instead of after every single number.
