# Methods: Passing and Returning References of an Object

Every method call so far has passed primitives like `int` and `double` as parameters. This lesson looks at what changes when you pass and return whole objects instead, and why "Java passes objects by reference" is a half-truth that trips people up on the AP exam.

## A Quick Reminder: Passing Primitives

When you pass a primitive to a method, the method gets its own copy of the value. Changes to that copy never reach the caller.

```python
public static void addTen(int x) {
    x = x + 10;
}

int num = 5;
addTen(num);
System.out.println(num); // still 5, addTen only changed its own local copy
```

## Passing an Object: Mutation Reaches the Caller

Objects behave differently, but not for the reason people usually assume. Java is still passing by value. What gets copied for an object argument is the *reference* (the address of the object), not the object itself. The method's copy of the reference points at the exact same object in memory as the caller's variable, so changes made through it are visible to both.

We'll use the `Shape` class from the last two lessons.

```python
public class Shape {
    protected String name;
    private int length;
    private int width;

    public Shape(String name, int length, int width) {
        this.name = name;
        this.length = length;
        this.width = width;
    }

    public int get_length() {
        return this.length;
    }

    public void set_length(int a) {
        this.length = a;
    }
}

public class ShapeTools {
    // Takes a reference to a Shape and mutates the object it points to
    public static void resize(Shape s) {
        s.set_length(100);
    }
}

Shape shape = new Shape("square", 4, 4);
ShapeTools.resize(shape);
System.out.println(shape.get_length()); // 100 -- the caller's object was mutated
```

## Reassignment Is Not Mutation

<div class="callout callout-hint">
<h4>Common Pitfall</h4>

Since the reference itself is passed *by value*, reassigning the parameter inside the method only redirects the method's own local copy. It does not change what the caller's variable points to.

</div>

Compare the mutation example above with this one, which only reassigns the parameter:

```python
public class Shape {
    protected String name;
    private int length;
    private int width;

    public Shape(String name, int length, int width) {
        this.name = name;
        this.length = length;
        this.width = width;
    }

    public String get_name() {
        return this.name;
    }
}

public class ShapeTools {
    // Reassigns the LOCAL parameter s to point at a brand new Shape.
    // This never touches the caller's variable.
    public static void reassign(Shape s) {
        s = new Shape("triangle", 3, 3);
    }
}

Shape shape2 = new Shape("square", 4, 4);
ShapeTools.reassign(shape2);
System.out.println(shape2.get_name()); // still "square"
```

## Returning Object References

A method can also return a reference to an object. That reference might point to a brand new object, or it might hand back a reference to an object that already existed. Either way, no copy of the object is made, just the reference.

One common pattern is a method returning `this`, a reference to the current object, so calls can be chained together.

```python
public class Counter {
    private int value = 0;

    // Returns a reference to this same Counter object, not a copy
    public Counter increment() {
        this.value++;
        return this;
    }

    public int getValue() {
        return this.value;
    }
}

Counter c = new Counter();
c.increment().increment().increment(); // chained, since each call returns the same object
System.out.println(c.getValue()); // 3
```

```python
public class Shape {
    protected String name;
    private int length;
    private int width;

    public Shape(String name, int length, int width) {
        this.name = name;
        this.length = length;
        this.width = width;
    }

    public String get_name() {
        return this.name;
    }

    public int get_length() {
        return this.length;
    }

    public void set_length(int a) {
        this.length = a;
    }
}

public class ShapeTools {
    public static void mutate(Shape s) {
        s.set_length(50);
    }
    public static void replace(Shape s) {
        s = new Shape("new", 1, 1);
    }
}

Shape original = new Shape("original", 4, 4);
ShapeTools.mutate(original);
ShapeTools.replace(original);
System.out.println(original.get_length());
System.out.println(original.get_name());
```

`50`, then `original`. `mutate` changes a field on the object `original` points to, so that change is visible after the call. `replace` only reassigns its own local copy of the reference to a new object, so `original` in the caller still points to the same object it always did.
