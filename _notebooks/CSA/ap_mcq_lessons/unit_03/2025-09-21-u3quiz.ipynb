### Question 5

A method receives a `Shape` object as a parameter and calls a setter on it that changes one of its fields. After the method returns, is that change visible to the code that called the method?<br>
a) No, because Java passes objects by value<br>
b) Yes, because the parameter is a copy of the reference, and that copy still points to the same object<br>
c) No, unless the method returns the object<br>
d) Yes, but only if the method is `static`

<details>
<summary>Show answer</summary>

b) Yes. Java always passes by value, but for an object argument, the value being copied is the reference itself. Both the caller's variable and the method's parameter point at the same object, so mutating it through either one is visible through both.

</details>

```python
// Q5 Hack: Predict the output, then run it to check.

public class Box {
    private int size;
    public Box(int size) { this.size = size; }
    public void setSize(int size) { this.size = size; }
    public int getSize() { return size; }
}

public class BoxTools {
    public static void mutate(Box b) {
        b.setSize(99);
    }
    public static void reassign(Box b) {
        b = new Box(1);
    }
}

Box box = new Box(10);
BoxTools.mutate(box);
BoxTools.reassign(box);
System.out.println(box.getSize());
```

### Question 7
