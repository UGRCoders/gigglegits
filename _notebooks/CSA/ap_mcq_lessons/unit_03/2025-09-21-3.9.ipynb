## Polymorphism

* "A reference variable is **polymorphic** when it can refer to objects of different classes in the code"
* "A method is **polymorphic** when it is overridden in at least one subclass"
* "**Polymorphism** is the act of executing an overridden ```non-static``` method from the correct class at runtime based on the actual object type"

````java
public class Shape {
    protected String name;
    private int length;
    private int width;

    // Parameterized constructor
    public Shape(String name, int length, int width) {
        this.name = name;
        this.length = length;
        this.width = width;
    }

    // Method to calculate the area
    public double calc_area() {
        return this.length * this.width;
    }
}

public class Triangle extends Shape {
    private int side1;
    private int side2;
    private int side3;

    // Constructor that takes a name and three side lengths
    public Triangle(String name, int s1, int s2, int s3) {
        super(name, 0, 0); // Call to Shape constructor to set the name
        this.name = "triangle";
        this.side1 = s1;
        this.side2 = s2;
        this.side3 = s3;
    }

    @Override
    public double calc_area() {
        double s = (side1 + side2 + side3) / 2.0; // Semi-perimeter
        return Math.sqrt(s * (s - side1) * (s - side2) * (s - side3));
    }
}

// declares a Shape-typed reference pointing to a new Triangle object
Shape triangle = new Triangle("Equilateral",1,1,1); 
// calc_area is overridden, so Triangle's version runs
System.out.println(triangle.calc_area()); 
````

## How Does This Work?

<img width="441" alt="Diagram of a Shape reference pointing to a Triangle object in memory" src="https://github.com/user-attachments/assets/11a11c1e-df70-4b9d-9d40-fc0720363d30">

Let's start with this line:
```java
Shape myObject = new Triangle();
```

It may look like you are creating a Shape object out of a Triangle, but you are not. You are creating a Triangle object. The difference is that `myObject` is declared as a Shape, so it can only see the Shape parts of that Triangle.

* A reference variable can store a reference to an object of its declared class, or to an object of any subclass of its declared class.

This also means that if the declared type is the superclass, and you try to call a method that only exists in the subclass, the code will not compile. You can fix this by casting the variable to the subclass type, since the object it actually points to is a subclass object. After casting, calling a subclass-only method works fine.

Now consider running an overridden method:
```java
myObject.calc_area(); // runs Triangle's calc_area, not Shape's
```

When you call a method on `myObject`, Java starts at the actual (Triangle) object and checks whether that class overrides the method. If it does not, Java moves up the inheritance chain until it either finds an override or reaches the original method in Shape.

## Static vs Dynamic Types

**static types**: static types is simply the type.

**dynamic types**: dynamic type is the type of the value during runtime

````java
class Shape {
    String getName(){
        return "Shape";
    }
}

class Square extends Shape{
    @Override
    String getName(){
        return "Square";
    }
}

Shape myShape = new Square();
````

<img width="643" alt="Diagram showing myShape's static type Shape versus its dynamic type Square" src="https://github.com/user-attachments/assets/2cf84906-8a22-4c0d-aa4f-096e85eb822c">

**Shape** is the static type, but at runtime ```myShape``` is referencing a **Square**, so ```myShape```'s dynamic type is **Square**.

## Compile-Time vs Run-Time methods

Compile time is when you are writing your code. Once you have written it, your code is *compiled* into something the computer can run.

Run time is when the code is actually running, after compilation is complete.

There are several differences between compile time and run time, and the one we care about here is method access. When you assign a Square to a Shape-typed variable, you cannot call methods that exist only on Square, you can only call methods that Shape declares. **But why?**

This comes down to how the compiler checks your code versus what actually exists in memory.

At compile time, Java only knows the *static* type of a variable, so it only allows method calls that the static type declares. It does not matter that the object referenced at run time is actually a Square with extra methods. The compiler never looks that far ahead.

So even though the object sitting in memory does have the Square-only methods, calling them through a Shape-typed variable is rejected before the program ever runs.

If you want to call a method that only exists on the subclass, you can use **down-casting** to tell the compiler to treat the variable as the subclass type.

````java
class Shape {
    String getName(){
        return "Shape";
    }
}

class Square extends Shape{
    @Override
    String getName(){
        return "Square";
    }

    int getSides(){
        return 4;
    }
}

Shape myShape = new Square(); // this does not have access to the Square methods despite referencing a Square

Square mySquare = (Square)myShape; // down-cast the Shape to a Square to run the Square specific methods
System.out.println(mySquare.getSides()); // after down-casting you can now run the Square methods
````

## Examples of Polymorphism and the effects
Here are some examples of Polymorphism.

1. Executing the overridden method from the referenced subclass in the datatype of the superclass.

````java
class Water {
    public String typeOfWater(){
        return "water";
    }
}

class Lake extends Water {
    @Override
    public String typeOfWater(){
        return "lake";
    }
}

// declares a Water-typed reference pointing to a new Lake object
Water lake = new Lake(); 
// typeOfWater is overridden, so Lake's version runs
System.out.println(lake.typeOfWater()); 
````

2. You can pass a subclass into an argument that is asking for the parent class.

````java
class Shape{
    public int getSize(){
        return 1;
    }
}

class Square extends Shape{
    @Override
    public int getSize(){
        return 2;
    }
}

int getSizePlusOne(Shape s){ //argument of class "Shape"
    return s.getSize() +1;
}

Shape myShape = new Shape();
//passes through a "Shape"
System.out.println(getSizePlusOne(myShape)); 

Square mySquare = new Square();
//passes through a "Square" as a "Shape" with the square's "getSize()" method
System.out.println(getSizePlusOne(mySquare)); 
````

3. You can cast from the superclass to the subclass (down-casting). The superclass must be referencing the subclass.

````java
class Electronic{
    public void playSound(){
        System.out.println("Beep boop");
    }
}

class Computer extends Electronic{
    @Override
    public void playSound(){
        System.out.println("Click clack");
    }
}

// declares an Electronic-typed reference pointing to a new Computer object
Electronic device = new Computer(); 

// down-casts the Electronic reference to a Computer reference
Computer computer = (Computer)device; 

computer.playSound();
````

````java
class Electronic{
    public void playSound(){
        System.out.println("Beep boop");
    }
}

class Computer extends Electronic{
    @Override
    public void playSound(){
        System.out.println("Click clack");
    }
}

Electronic device = new Electronic(); // creates an Electronic object, not a Computer

Computer computer = (Computer)device; // compiles, but throws ClassCastException at runtime because device isn't actually a Computer
````

4. Down-casting doesn't mean changing the reference to the object, so polymorphism will still apply.

````java
class Furniture{
    String getName(){
        return "Furniture";
    }
}

class Table extends Furniture{
    @Override
    String getName(){
        return "Table";
    }
}

class CoffeeTable extends Table{
    @Override
    String getName(){
        return super.getName() + " Coffee";
    }
}

Furniture myTable = new CoffeeTable();
// runs CoffeeTable's method
System.out.println(myTable.getName());

Table myOtherTable = (Table)myTable;
// the method that runs doesn't change just because we cast down the reference
System.out.println(myOtherTable.getName());
````

5. You can cast upward and polymorphic behavior will apply.

````java
class Sport{
    public int numberOfPlayers(){
        return (int) Double.NaN; // casting NaN to int yields 0 in Java
    }
    public String name(){
        return "Sport";
    }
}

class Soccer extends Sport{
    @Override
    public int numberOfPlayers(){
        return 11;
    }
    @Override
    public String name(){
        return "Soccer";
    }
}

// creates a Soccer object
Soccer mySoccer = new Soccer();

Sport mySoccerSport = (Sport)mySoccer;
System.out.println(mySoccerSport.numberOfPlayers());
````

### Polymorphism with the Object Class

Polymorphism also applies with the Object superclass. Technically, any class or object is an `Object`.

````java
class Fruit{
    @Override
    public String toString(){
        return "I'm a Fruit!";
    }
}
class Banana extends Fruit{
    @Override
    public boolean equals(Object obj){ // overrides equals from Object
        return true;
    }
}


Object banana = new Banana();
System.out.println(banana.toString());
// if equals() were not overridden from Object, this would always return false
System.out.println(banana.equals(null)); 
````
