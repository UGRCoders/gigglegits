# Class Variables and Methods

Not everything a class needs belongs to one object. This lesson covers `static`: fields and methods that belong to the class itself, shared across every instance instead of copied into each one.

## Instance Variables vs. Class Variables

Every `Shape` we've made so far gets its own `name`, `length`, and `width`. Those are instance variables: each object holds its own separate copy. A variable marked `static` works differently. There is exactly one copy of it, and it belongs to the class, not to any single object. Every instance shares that same copy.

A natural use for this is counting how many `Shape` objects have been created.

```python
public class Shape {
    protected String name;
    private int length;
    private int width;
    private static int shapeCount = 0; // one copy shared by every Shape

    public Shape(String name, int length, int width) {
        this.name = name;
        this.length = length;
        this.width = width;
        shapeCount++; // every constructor call increments the shared counter
    }

    public static int getShapeCount() {
        return shapeCount;
    }
}

Shape s1 = new Shape("square", 4, 4);
Shape s2 = new Shape("rectangle", 3, 6);
Shape s3 = new Shape("triangle", 2, 5);

System.out.println("Shapes created: " + Shape.getShapeCount()); // 3, called on the class itself
```

Notice `Shape.getShapeCount()` is called on the class name `Shape`, not on any particular shape object. That's the giveaway for a static method: you're asking the class a question, not asking a specific object. You've been calling static methods since Unit 1 without necessarily naming them that: `Math.random()` and `Math.max()` are both static methods on the `Math` class.

## Why Static Methods Can't Use `this`

A static method runs without any particular object behind it, so there is no object for `this` to refer to. That also means a static method cannot directly touch instance variables, since instance variables only exist once an object has been created.

```python
// this will not compile
public class Shape {
    private int length; // instance variable, belongs to a specific object

    public static int getLength() {
        return length; // error: non-static variable length cannot be referenced from a static context
    }
}
```

Since `getLength()` is static, Java could call it before any `Shape` object exists at all. At that point there is no `length` field to return, since `length` only exists once you've built a specific `Shape`. `shapeCount` worked fine as static because it isn't tied to any one object in the first place.

## When To Use static

- Use `static` for data or behavior that describes the class as a whole: a running total, a shared constant, a utility calculation that doesn't need any object's state (like `Math.max`).
- Use an instance variable or method for anything that varies from object to object, like a `Shape`'s own `length` and `width`.
