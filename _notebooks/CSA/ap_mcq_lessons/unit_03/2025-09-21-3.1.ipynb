# Abstraction and Program Design

Every class you write is an act of abstraction: you decide what the rest of your program needs to know about something, and you hide everything else. This lesson is about that decision, before we get into the Java syntax for building classes.

## Why Bother With Classes?

Before you learn the syntax for writing a class, it helps to see the problem classes actually solve. Say you need to track a handful of students and their grades. Without classes, you might reach for a set of parallel arrays.

````python
import java.util.ArrayList;

ArrayList<String> names = new ArrayList<>();
ArrayList<Double> averages = new ArrayList<>();

names.add("Amara");
averages.add(91.5);
names.add("Devon");
averages.add(88.0);

// Someone later decides to remove a student who dropped the class...
names.remove(0); // removes "Amara" from names
// ...but forgets averages has to be updated too.

for (int i = 0; i < names.size(); i++) {
    System.out.println(names.get(i) + ": " + averages.get(i));
}
// Prints "Devon: 91.5" -- Devon now has Amara's average. The two arrays
// only stayed in sync as long as every piece of code that touched one
// remembered to touch the other. Nothing in the language enforces that.
````

Nothing about that bug is a typo or a syntax error. It is a design problem: a student's name and average are two separate facts about the same student, but the program has no way to say that. This is exactly what a class fixes. A `Student` object bundles a name with its own average, so they cannot drift apart as long as you work through the object.

```java
public class Student {
    private String name;
    private double average;

    public Student(String name, double average) {
        this.name = name;
        this.average = average;
    }

    public String getName() {
        return name;
    }

    public double getAverage() {
        return average;
    }
}
```

Now a list of `Student` objects can gain or lose entries and each name always carries its own average with it. There is no second array to forget about.

## Abstraction: Interface vs. Implementation

Abstraction means hiding how something works and exposing only what it does. When you press the volume button on a TV remote, you do not need to know how the infrared signal is encoded. The button is the interface. The signal encoding is the implementation.

A class works the same way:

- **Interface**: the public methods a class exposes, the promises it makes to anyone using it (for example, `getAverage()` always returns a `Student`'s average).
- **Implementation**: the private fields and logic behind those methods, which the rest of the program should never need to touch directly.

You already rely on this every time you call `Math.random()`. You know it returns a `double` between 0.0 and 1.0. You do not know, and do not need to know, exactly how the number is generated internally. If that internal algorithm changed tomorrow, every program calling `Math.random()` would keep working without a single edit, because the interface never changed.

## Designing a Class: A Simple Process

Before writing any Java, it helps to answer three questions on paper:

1. **What thing am I modeling?** Pick a noun: a `Student`, a `Shape`, a `BankAccount`.
2. **What does it need to remember?** These become instance variables, for example a `Student`'s name and average.
3. **What does it need to do?** These become methods, for example `getAverage()` or `addGrade(double grade)`.

Once you have those three answers, you can also decide what should be part of the public interface and what should stay hidden as an implementation detail.

<h4>Common Pitfall</h4>

New programmers often make every field `public` "just in case something needs it." Good abstraction does the opposite: hide fields and expose only the methods a user of the class actually needs. The next few lessons build up to exactly how to lock this down with constructors and access modifiers.
