# Constructors

A constructor is the code that runs the moment you write `new`. This lesson covers how to write one, why Java gives you a free one until you don't, and how a class can offer more than one way to build an object.

<div class="callout callout-objectives">
<h4>Objectives</h4>

- Write a parameterized constructor that initializes instance variables
- Explain why Java's implicit default constructor disappears once you write your own constructor
- Overload constructors so a class can be built more than one way
- Recognize the common mistake of giving a constructor a return type

</div>

## What Is a Constructor?

A constructor is a special block of code that shares its name with the class and has no return type, not even `void`. It runs automatically whenever `new` creates an object, and its job is to set that object's instance variables to sensible starting values.

Let's build up a `Shape` class one piece at a time, starting with a single parameterized constructor.

````python
public class Shape {
    protected String name;
    private int length;
    private int width;

    // Parameterized constructor
    public Shape(String name, int length, int width) {
        this.name = name;
        this.length = length;
        this.width = width;
    }

    public void print_shape() {
        System.out.println("Shape: " + this.name + ", length=" + this.length + ", width=" + this.width);
    }
}

Shape s1 = new Shape("rectangle", 4, 6);
s1.print_shape();
````

Notice `this.name = name;`. The constructor's parameter is also called `name`, so inside the constructor's body the plain word `name` refers to the parameter, not the field. Writing `this.name` explicitly says "the field that belongs to this object," which is what breaks the tie. Without `this`, the line `name = name;` would just assign the parameter to itself and the field would stay unset.

## Default Constructors

What happens if you call `new Shape()` with no arguments? Run the cell below and see the actual compiler error:

````python
Shape s2 = new Shape(); // does NOT compile: Shape has no no-argument constructor
````

That fails, and this is the pitfall to watch for. Java only supplies a free no-argument constructor automatically if your class defines **no constructors at all**. The moment you write even one constructor of your own, Java stops supplying that implicit one. If you still want a no-argument option, you have to write it yourself.

## Constructor Overloading

A class can have more than one constructor, as long as each one has a different parameter list. This is called overloading, and it lets a class be built more than one way. Here is the full `Shape` class with both a no-argument constructor and a parameterized one, plus getters and setters so other code can read and update its fields. This is the same `Shape` class the next lesson picks up and builds on.

````python
public class Shape {
    protected String name;
    private int length;
    private int width;

    // Default constructor
    public Shape() {
        this.name = "shape";
        this.length = 10;
        this.width = 5;
    }

    // Parameterized constructor
    public Shape(String name, int length, int width) {
        this.name = name;
        this.length = length;
        this.width = width;
    }

    // Getter methods
    public String get_name() {
        return this.name;
    }

    public int get_length() {
        return this.length;
    }

    public int get_width() {
        return this.width;
    }

    // Setter methods
    public void set_name(String n) {
        this.name = n;
    }

    public void set_length(int a) {
        this.length = a;
    }

    public void set_width(int b) {
        this.width = b;
    }

    // Method to calculate the area
    public double calc_area() {
        return this.length * this.width;
    }

    // Method to calculate the perimeter
    public double calc_perimeter(){
        return 2*this.length + 2*this.width;
    }

    // Method to print the shape
    public void print_shape() {
        System.out.println("Shape: " + this.name);
    }

    // Additional method to print something
    public void print_something() {
        System.out.println("This is a shape");
    }
}

Shape defaultShape = new Shape();
Shape customShape = new Shape("rectangle", 4, 6);
defaultShape.print_shape();
customShape.print_shape();
System.out.println("Default area: " + defaultShape.calc_area());
System.out.println("Custom area: " + customShape.calc_area());
````

<div class="callout callout-hint">
<h4>Common Pitfall</h4>

A constructor with a return type, even `void`, is not a constructor anymore. It becomes a regular method that happens to share the class's name.

```java
public class Shape {
    protected String name;

    public void Shape(String name) { // this is a METHOD, not a constructor
        this.name = name;
    }
}
```

`new Shape("triangle")` will not call that method. Java will look for an actual no-argument constructor instead (and fail to compile if there isn't one), because the `void Shape(String name)` above is just a method that happens to be named the same as the class.

</div>
