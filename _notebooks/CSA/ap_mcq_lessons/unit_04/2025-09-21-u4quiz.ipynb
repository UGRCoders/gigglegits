### Question 3

```java
int[] values = {10, 20, 30};
for (int i = 0; i <= values.length; i++) {
    System.out.println(values[i]);
}
```

What happens when this code runs?<br>
a) It prints 10, 20, 30 and stops cleanly<br>
b) It throws an ArrayIndexOutOfBoundsException<br>
c) It prints 10, 20, 30, 0<br>
d) It's a compile error

<details>
<summary>Show answer</summary>

b) The condition uses `<=` instead of `<`. When `i` reaches `values.length` (3), `values[3]` doesn't exist, since the valid indices are only 0 through `length - 1`.

</details>

### Question 4

A linear search method searches an array for a target value and doesn't find it anywhere. By convention, what should it return?<br>
a) `0`<br>
b) `null`<br>
c) `-1`<br>
d) The array's length

<details>
<summary>Show answer</summary>

c) `-1`. Since `-1` can never be a valid array index, it's an unambiguous signal that no match was found, unlike `0`, which could also be a real, valid index.

</details>

### Question 9

Design and implement a `GradeBook` class that pulls together several ideas from this unit: parallel data for multiple students, an ArrayList, a 2D array, and a search or sort algorithm.

Required elements:
- An `ArrayList<String>` field `students`, holding student names.
- A 2D array field `int[][] quizScores`, where `quizScores[i]` holds every quiz score for `students.get(i)`, in the order the quizzes were given. (Every student has taken the same number of quizzes.)
- A constructor that takes the number of students and the number of quizzes, and initializes both fields (an empty `students` list, and a `quizScores` array sized for that many students and quizzes, with every score starting at 0).
- A method `addStudent(String name)` that adds a name to `students`.
- A method `recordScore(int studentIndex, int quizIndex, int score)` that stores a score in `quizScores` at the right position.
- A method `averageForStudent(int studentIndex)` that traverses one student's row and returns their average score across all quizzes.
- A method `highestScoringStudent()` that traverses the entire 2D array, computes each student's average, and returns the name of the student with the highest average.

Think about which parts of this need a 1D traversal, which need a 2D traversal, and where you're reading from the ArrayList versus the 2D array before you start writing code.

<details>
<summary>Show sample answer</summary>

```java
import java.util.ArrayList;

public class GradeBook {
    private ArrayList<String> students;
    private int[][] quizScores;

    public GradeBook(int numStudents, int numQuizzes) {
        students = new ArrayList<>();
        quizScores = new int[numStudents][numQuizzes];
    }

    public void addStudent(String name) {
        students.add(name);
    }

    public void recordScore(int studentIndex, int quizIndex, int score) {
        quizScores[studentIndex][quizIndex] = score;
    }

    public double averageForStudent(int studentIndex) {
        int total = 0;
        for (int score : quizScores[studentIndex]) {
            total += score;
        }
        return total / (double) quizScores[studentIndex].length;
    }

    public String highestScoringStudent() {
        int bestIndex = 0;
        double bestAverage = averageForStudent(0);

        for (int i = 1; i < students.size(); i++) {
            double average = averageForStudent(i);
            if (average > bestAverage) {
                bestAverage = average;
                bestIndex = i;
            }
        }
        return students.get(bestIndex);
    }

    public static void main(String[] args) {
        GradeBook book = new GradeBook(2, 3);
        book.addStudent("Amy");
        book.addStudent("Bo");

        book.recordScore(0, 0, 90);
        book.recordScore(0, 1, 85);
        book.recordScore(0, 2, 95);

        book.recordScore(1, 0, 70);
        book.recordScore(1, 1, 100);
        book.recordScore(1, 2, 80);

        System.out.println(book.averageForStudent(0));
        System.out.println(book.highestScoringStudent());
    }
}
GradeBook.main(null)
```

</details>

````python
// Q4 Hack: write indexOf(int[] values, int target), returning the index of
// the first match, or -1 if target isn't in the array.
public static int indexOf(int[] values, int target) {
    for (int i = 0; i < values.length; i++) {
        if (values[i] == target) {
            return i;
        }
    }
    return -1;
}

int[] nums = {4, 8, 15, 16, 23, 42};
System.out.println(indexOf(nums, 16));
System.out.println(indexOf(nums, 99));
````
