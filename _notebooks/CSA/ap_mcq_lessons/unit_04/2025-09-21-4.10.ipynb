# Implementing ArrayList Algorithms

Every algorithm from Unit 4's array lessons, search, sort, sum, has an ArrayList equivalent. The logic is identical, only the syntax for reading and writing an element changes: `.get(i)` instead of `[i]`, `.set(i, value)` instead of `[i] =`, and `.size()` instead of `.length`.

## Linear Search on an ArrayList

````python
import java.util.ArrayList;

public class ListSearch {
    public static int indexOf(ArrayList<Integer> values, int target) {
        for (int i = 0; i < values.size(); i++) {
            if (values.get(i) == target) {
                return i;
            }
        }
        return -1;
    }

    public static void main(String[] args) {
        ArrayList<Integer> scores = new ArrayList<>();
        scores.add(85);
        scores.add(91);
        scores.add(76);
        scores.add(100);

        System.out.println(indexOf(scores, 76));  // 2
        System.out.println(indexOf(scores, 60));   // -1
    }
}
ListSearch.main(null);
````

<div class="callout callout-hint">
<h4>Attention: Common Mistake</h4>

`values.get(i) == target` works here because `Integer` values in the range -128 to 127 are cached and unboxed automatically when compared to a primitive `int`, so `==` behaves like value comparison. That's a narrow, easy-to-get-wrong special case. For anything that isn't a small cached integer, most of all for `String` or any other object type, use `.equals()` instead of `==` to compare an element you got from an ArrayList.

</div>

## Selection Sort on an ArrayList

Same algorithm as the array version from the last lesson: find the smallest remaining value, swap it into place, move on.

````python
import java.util.ArrayList;

public class ListSelectionSort {
    public static void sort(ArrayList<Integer> values) {
        for (int outer = 0; outer < values.size() - 1; outer++) {
            int minIndex = outer;
            for (int inner = outer + 1; inner < values.size(); inner++) {
                if (values.get(inner) < values.get(minIndex)) {
                    minIndex = inner;
                }
            }
            if (minIndex != outer) {
                int temp = values.get(outer);
                values.set(outer, values.get(minIndex));
                values.set(minIndex, temp);
            }
        }
    }

    public static void main(String[] args) {
        ArrayList<Integer> scores = new ArrayList<>();
        scores.add(85);
        scores.add(91);
        scores.add(76);
        scores.add(100);
        scores.add(88);

        sort(scores);
        System.out.println(scores); // [76, 85, 88, 91, 100]
    }
}
ListSelectionSort.main(null);
````

## Insertion Sort on an ArrayList

Insertion sort builds a sorted section at the front of the list one element at a time. It takes the next unsorted value and slides it backward past every already-sorted value that's bigger than it, until it lands in the right spot.

````python
import java.util.ArrayList;

public class ListInsertionSort {
    public static void sort(ArrayList<Integer> values) {
        for (int outer = 1; outer < values.size(); outer++) {
            int toInsert = values.get(outer);
            int inner = outer - 1;

            while (inner >= 0 && values.get(inner) > toInsert) {
                values.set(inner + 1, values.get(inner));
                inner--;
            }
            values.set(inner + 1, toInsert);
        }
    }

    public static void main(String[] args) {
        ArrayList<Integer> scores = new ArrayList<>();
        scores.add(85);
        scores.add(91);
        scores.add(76);
        scores.add(100);
        scores.add(88);

        sort(scores);
        System.out.println(scores); // [76, 85, 88, 91, 100]
    }
}
ListInsertionSort.main(null);
````

`toInsert` is saved before the `while` loop starts, since the loop is about to overwrite `values.get(outer)`'s old position while shifting everything else right. The `while` loop keeps sliding sorted values one spot to the right as long as they're bigger than `toInsert`, then `toInsert` gets placed into the gap that opened up right after the loop stops.

<div class="callout callout-practice">
<h4>Popcorn Hack</h4>

Both sorts finalize one element per pass through the outer loop, but they finalize different ends of the list. Selection sort finalizes the front of the array (the smallest unplaced value gets locked in at `outer`). Where does insertion sort's sorted section live, and does it grow from the front or by absorbing the next element each pass?

<details>
<summary>Show answer</summary>

Insertion sort's sorted section is always `values[0]` through `values[outer - 1]`, and it grows by one from the front each pass, since `outer` absorbs the next unsorted element into the already-sorted front section on every iteration. Selection sort instead searches the entire unsorted remainder every pass to find the correct next value; insertion sort just takes whatever's next and finds where it belongs.

</details>
