# Introduction to Using Data Sets

Real programs rarely work with just one piece of data. A weather app doesn't track one temperature, it tracks a week of them. A grade tracker doesn't hold one score, it holds every score in the class. This lesson looks at the problem that creates, and sets up why the rest of this unit exists.

## What Is a Data Set

A data set is just a collection of related values: a week of temperatures, every student's grade on a quiz, the price of every item in a store. What makes it a data set rather than a pile of unrelated variables is that every value in it is the same kind of thing, and you usually want to do the same operation to all of them: find the average, find the highest, count how many meet some condition.

## The Problem With Separate Variables

Here's how you'd store five quiz scores using what you already know from Unit 1:

```java
int score1 = 85;
int score2 = 91;
int score3 = 76;
int score4 = 100;
int score5 = 88;

int total = score1 + score2 + score3 + score4 + score5;
double average = total / 5.0;
System.out.println(average);
```

This works, technically, but it falls apart fast:

- What if the class has 30 students, not 5? You'd need 30 separate variable names.
- What if you don't know in advance how many scores there will be, because students are still submitting?
- What if you need to find the highest score? You'd have to write a chain of comparisons between every single variable.
- What if you need to loop through "all the scores"? You can't loop over `score1`, `score2`, `score3`... a loop needs to be able to say "give me the next one," and separate variables have no next.

Every one of these problems has the same root cause: the scores are related to each other, but the code has no way to represent that relationship. Each `scoreN` variable is its own isolated box with its own name, even though conceptually they're all "the scores."

## What a Data Set Needs

For a group of values to be treated as one data set that a program can process uniformly, two things need to be true:

1. **Same type.** Every element has to be the same kind of value, so the same operations (`+`, comparisons, method calls) make sense on all of them. You can average a collection of `double` values; you can't meaningfully average a collection that mixes numbers and strings.
2. **Ordered access.** You need a way to ask for "the next one" or "the one at position `i`," so a loop can visit every element without you writing out each variable name by hand.

## Where This Is Going

Java's answer to this problem is a single variable that holds many values of the same type, accessed by position instead of by a unique name for each one. That's exactly what an **array** is, and it's the subject of the next lesson. Instead of `score1` through `score5`, you'll write one variable, `scores`, that holds all five, and a loop that can visit every element regardless of how many there are.
