# Using Text Files

Every array you've built so far has been typed directly into the source code. Real data usually doesn't start out that way, it lives in a file. This lesson covers reading a text file into your program and writing data back out to one.

<div class="callout callout-objectives">
<h4>Objectives</h4>

- Read a text file line by line using `Scanner` and `File`
- Handle the checked exception a missing or unreadable file throws
- Load the lines of a file into an array or ArrayList for further processing
- Write data to a text file using `PrintWriter`

</div>

## Reading a File with Scanner

You've already used `Scanner` to read keyboard input, `new Scanner(System.in)`. Point it at a `File` instead, and it reads from that file the same way, one token or one line at a time.

```java
import java.io.File;
import java.io.FileNotFoundException;
import java.util.Scanner;

public class ReadFile {
    public static void main(String[] args) {
        try {
            File file = new File("scores.txt");
            Scanner input = new Scanner(file);

            while (input.hasNextLine()) {
                String line = input.nextLine();
                System.out.println(line);
            }

            input.close();
        } catch (FileNotFoundException e) {
            System.out.println("Could not find scores.txt");
        }
    }
}
```

`FileNotFoundException` is a **checked exception**: the compiler forces you to either catch it or declare that your method throws it, because a missing file is an entirely normal, expected thing to happen, not a bug. `input.hasNextLine()` is what makes this a traversal of the file: it returns `false` once there's nothing left, so the loop stops itself instead of you having to know the line count in advance.

## Loading a File Into a Collection

Printing each line as you read it is fine, but usually you want the file's contents available afterward, as a data set you can traverse, search, or sort like anything else in this unit. An ArrayList is a natural fit here, since you don't know the line count ahead of time.

```java
import java.io.File;
import java.io.FileNotFoundException;
import java.util.ArrayList;
import java.util.Scanner;

public class LoadNames {
    public static void main(String[] args) {
        ArrayList<String> names = new ArrayList<>();

        try {
            File file = new File("names.txt");
            Scanner input = new Scanner(file);

            while (input.hasNextLine()) {
                names.add(input.nextLine());
            }

            input.close();
        } catch (FileNotFoundException e) {
            System.out.println("Could not find names.txt");
        }

        System.out.println("Loaded " + names.size() + " names.");
        for (String name : names) {
            System.out.println(name);
        }
    }
}
```

Once the file is loaded into `names`, everything from the last two lessons applies to it: you can linear search it, sort it, count matches, anything you'd do to a data set that started as an array literal.

<div class="callout callout-hint">
<h4>Attention: Common Mistake</h4>

Mixing `nextLine()` with `nextInt()` or `next()` on the same `Scanner` is a classic bug: `nextInt()` reads the number but leaves the trailing newline character in the buffer, so the very next `nextLine()` call reads an empty string instead of the line you expected. If a file has numbers and text mixed together, either read everything with `nextLine()` and parse each line yourself, or call an extra `nextLine()` to consume the leftover newline after a `nextInt()`.

</div>

## Writing to a File

`PrintWriter` writes to a file the same way `System.out` prints to the console, using `.println()`.

```java
import java.io.FileNotFoundException;
import java.io.PrintWriter;

public class WriteFile {
    public static void main(String[] args) {
        try {
            PrintWriter output = new PrintWriter("results.txt");

            output.println("Total: 440");
            output.println("Average: 88.0");

            output.close();
        } catch (FileNotFoundException e) {
            System.out.println("Could not create results.txt");
        }
    }
}
```

<div class="callout callout-note">
<h4>Note</h4>

Forgetting to `close()` a `Scanner` or a `PrintWriter` is a common source of bugs: with `PrintWriter` especially, output can sit in a buffer and never actually get written to disk until the file is closed. Always close what you open.

</div>
