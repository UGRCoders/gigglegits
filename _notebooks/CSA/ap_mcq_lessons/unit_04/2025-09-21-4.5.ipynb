## Sum, Average, and Counting

The pattern is always the same: declare a variable to hold the running result before the loop starts, update it once per element inside the loop, then use the final value after the loop ends.

```java
public class ArraySummary {
    public static void main(String[] args) {
        int[] scores = {85, 91, 76, 100, 88};

        int total = 0;
        for (int score : scores) {
            total += score;
        }
        double average = total / (double) scores.length;

        int passingCount = 0;
        for (int score : scores) {
            if (score >= 70) {
                passingCount++;
            }
        }

        System.out.println("Total: " + total);
        System.out.println("Average: " + average);
        System.out.println("Passing: " + passingCount);
    }
}
ArraySummary.main(null)
```

<div class="callout callout-hint">
<h4>Attention: Common Mistake</h4>

`total / scores.length` with two `int` operands performs integer division and truncates. That's why the average above is computed as `total / (double) scores.length`: casting one operand to `double` forces the whole division to happen in floating point. This is the same integer division rule from Unit 1, it just shows up constantly once you're averaging real data.

</div>

## Linear Search

A linear search checks each element in order until it finds a match or runs out of elements. By convention, it returns the index of the match, or `-1` if nothing matched, since `-1` can never be a valid array index.

```java
public class LinearSearch {
    public static int indexOf(int[] values, int target) {
        for (int i = 0; i < values.length; i++) {
            if (values[i] == target) {
                return i;
            }
        }
        return -1;
    }

    public static void main(String[] args) {
        int[] scores = {85, 91, 76, 100, 88};
        System.out.println(indexOf(scores, 100)); // 3
        System.out.println(indexOf(scores, 60));  // -1, not found
    }
}
LinearSearch.main(null)
```

Returning from inside the loop the moment a match is found is what makes this a search and not just a traversal: you stop as soon as you have your answer, instead of always checking every element.

## Selection Sort

Selection sort repeatedly finds the smallest remaining value and swaps it into place. The array ends up sorted from the front, one correct position at a time.

```java
public class SelectionSort {
    public static void sort(int[] values) {
        for (int outer = 0; outer < values.length - 1; outer++) {
            int minIndex = outer;
            for (int inner = outer + 1; inner < values.length; inner++) {
                if (values[inner] < values[minIndex]) {
                    minIndex = inner;
                }
            }
            if (minIndex != outer) {
                int temp = values[outer];
                values[outer] = values[minIndex];
                values[minIndex] = temp;
            }
        }
    }

    public static void main(String[] args) {
        int[] scores = {85, 91, 76, 100, 88};
        sort(scores);
        for (int score : scores) {
            System.out.print(score + " ");
        }
    }
}
SelectionSort.main(null)
```

Walking through it: `outer` marks the position that's about to be finalized. The inner loop scans everything from `outer + 1` to the end, looking for the smallest value in that unsorted remainder, and tracks its index in `minIndex`. After the inner loop finishes, one swap puts that smallest value into position `outer`. That's why the outer loop only needs to run to `values.length - 1`: once every position except the last one is finalized, the last element has nowhere else to go, it's already the largest, and must already be in the correct spot.

<div class="callout callout-hint">
<h4>Attention: Common Mistake</h4>

The swap needs a temporary variable because a Java variable can only hold one value at a time. Writing `values[outer] = values[minIndex]; values[minIndex] = values[outer];` overwrites `values[outer]` on the first line, so the second line copies that same new value right back, and the original value at `values[outer]` is gone. Save it in `temp` first.

</div>
