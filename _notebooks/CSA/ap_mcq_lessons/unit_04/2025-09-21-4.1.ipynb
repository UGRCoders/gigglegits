# Ethical and Social Issues Around Data Collection

Before this unit teaches you how to store and process data with arrays, it's worth stopping to ask a different question: what happens to people when software collects their data, and what responsibility does the programmer who wrote that software have?

<div class="callout callout-objectives">
<h4>Objectives</h4>

- Identify what counts as personal data and why some of it is more sensitive than other data
- Explain the difference between data collection with consent and data collection without it
- Describe how bias enters a data set, and how a biased data set produces a biased program
- List concrete ways a programmer can reduce the privacy risk of the software they build

</div>

## Data Is Personal Before It's a Data Structure

Every array and ArrayList you build in this unit will eventually hold something. Sometimes that's just numbers for a math problem. But a huge amount of real software exists specifically to collect, store, and process information about actual people: their location, their grades, their purchases, their messages, their health.

It's easy to think of "the data" as an abstract thing once it's sitting in a variable named `userData`. It stops feeling abstract the moment it's your own grades, your own location history, or a photo of your own face. Writing code that touches personal data comes with responsibility that writing code that sums an array of test scores doesn't.

## What Counts as Personal Data

Some categories of data are more sensitive than others, and treating all data the same way is itself a common mistake.

| Category | Examples | Why it matters |
| --- | --- | --- |
| Identifying | Name, student ID, email | Directly points back to one person |
| Behavioral | App usage, location history, browsing history | Reveals patterns and habits, even without a name attached |
| Sensitive | Health records, financial data, biometric data (face, fingerprint) | Causes serious harm if leaked or misused |
| Inferred | Predictions a system makes about you from other data | Can be wrong, and you often don't know it exists |

That last category, inferred data, is easy to overlook. A system doesn't need you to type in your interests if it can guess them from what you click on. Software can generate new personal data about you that you never directly gave it.

## Consent: Did Anyone Actually Agree to This?

Consent means the person the data is about knowingly agreed to it being collected, and understood roughly what it would be used for. A lot of real-world data collection fails this test in quiet ways:

- A privacy policy that's 40 pages of legal text nobody reads before clicking "I agree" is technically consent, but not meaningfully informed consent.
- An app that collects your location "to improve the experience" without saying it also sells that location data to advertisers is not being honest about what the consent covers.
- A school system, a fitness tracker, or a game that logs everything you do by default, with no way to opt out, isn't asking for consent at all.

<div class="callout callout-note">
<h4>Note</h4>

Consent isn't just a legal checkbox. As a programmer, you decide what data your program actually collects, and that decision happens before any user ever sees a privacy policy.

</div>

## Bias: Garbage In, Biased Out

A program that makes decisions, who gets a loan, what content gets recommended, who a facial recognition system flags as a match, is only as fair as the data it was trained or built on. If the historical data used to build a system reflects existing inequality, the program will learn to reproduce that inequality, often while looking perfectly "neutral" from the outside.

A well known real example: early facial recognition systems were often trained mostly on photos of lighter-skinned faces, which measurably made them less accurate at correctly identifying people with darker skin. The code wasn't written to be racist. The data set it learned from just wasn't representative of everyone it would eventually be used on. The harm still happened.

This matters for you as someone who will soon be building programs that process arrays and ArrayLists full of real values: the data set you choose to build or test with shapes what your program actually does, in ways that aren't always visible just by reading the code.

## What a Programmer Can Actually Do

You're not just a bystander to these problems. Concrete habits that reduce privacy risk in software you write:

- **Collect less.** If your program doesn't need a piece of data to function, don't collect it. Data you never collected can't be leaked.
- **Delete when you're done.** Don't keep personal data around indefinitely just because storage is cheap. Delete it once its purpose is served.
- **Anonymize where possible.** Strip out or hash identifying fields (like running a value through `.hashCode()`) when you only need the pattern in the data, not who it belongs to.
- **Be honest about what you collect.** If your program logs something, a user should be able to find out that it does.
- **Test with representative data.** If you're building something that makes decisions about people, check whether your test data actually reflects the range of people who will use it.
