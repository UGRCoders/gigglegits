The last lesson covered declaring, initializing, accessing, and updating a 2D array. Traversing means visiting every element, and since a 2D array is an array of arrays, that takes a loop inside a loop: a nested loop.

## Traversing

To traverse a 2D array, you need an outer loop and an inner loop: the outer loop moves through the rows, and for each row, the inner loop moves through every column in that row.

### Nested For Loop

A nested for loop is a loop with another loop inside its body. The outer loop's variable typically tracks the row, and the inner loop's variable tracks the column.

```java
public class NestedLoops {
    public static void main(String[] args) {
        for (int outer = 1; outer < 5; outer++) {
            for (int inner = 1; inner < 3; inner++) {
                System.out.print(inner + " ");
            }
            System.out.println();
        }
    }
}
NestedLoops.main(null)
```

For every single pass of the outer loop, the entire inner loop runs start to finish. That's why this prints the same two numbers, four times.

### Enhanced For Loop (Review)

Recall that an enhanced for loop traverses a 1D array without needing an index:

```java
public class EnhancedForLoopExample {
    public static void main(String[] args) {
        String[] fruits = {"Apple", "Banana", "Cherry", "Date", "Fig"};

        for (String fruit : fruits) {
            System.out.println(fruit);
        }
    }
}
EnhancedForLoopExample.main(null)
```

You can nest enhanced for loops the same way you nest standard for loops, which is exactly what you need for a 2D array: an enhanced for loop over the rows, with an enhanced for loop over each row's elements inside it.

## Nested Loops Traversing a 2D Array

```java
public class TwoDArrayTraversal {
    public static void main(String[] args) {
        int[][] twoDArray = {
            {1, 2, 3},
            {4, 5, 6},
            {7, 8, 9}
        };

        // Nested loops traverse row by row
        for (int row = 0; row < 3; row++) {
            for (int col = 0; col < 3; col++) {
                int element = twoDArray[row][col];
                System.out.print(element + " ");
            }
            System.out.println();
        }
    }
}
TwoDArrayTraversal.main(null)
```

## Using Nested Enhanced For Loops to Traverse a 2D Array

```java
public class EnhancedForLoop2DArray {
    public static void main(String[] args) {
        int[][] twoDArray = {
            {1, 2, 3},
            {4, 5, 6},
            {7, 8, 9}
        };

        // The outer loop pulls out each row (itself a 1D array)
        for (int[] row : twoDArray) {
            // The inner loop pulls out each element of that row
            for (int element : row) {
                System.out.print(element + " ");
            }
            System.out.println();
        }
    }
}
EnhancedForLoop2DArray.main(null)
```

## Row Major Order vs Column Major Order

<div class="callout callout-practice">
<h4>Popcorn Hack</h4>

</div>

<details>
<summary>Show answer</summary>

Row major order finishes an entire row before moving to the next row, meaning the outer loop is rows and the inner loop is columns. Column major order finishes an entire column before moving to the next column, meaning the outer loop is columns and the inner loop is rows. The code below makes this concrete.

</details>

Notice how switching which loop is on the outside changes the order elements are printed in:

```java
public class PrintArray {
    public static void printArray(String[][] grid) {
        // Row major: walk each row left to right, top to bottom
        for (int row = 0; row < grid.length; row++) {
            for (int col = 0; col < grid[0].length; col++) {
                System.out.print(grid[row][col] + " ");
            }
            System.out.println();
        }
    }

    public static void main(String[] args) {
        String[][] myGrid = {
            {"A", "B", "C"},
            {"D", "E", "F"},
            {"G", "H", "I"}
        };
        printArray(myGrid);
    }
}
PrintArray.main(null)
```

```java
public class PrintArray {
    public static void printArray(String[][] grid) {
        // Column major: the outer loop is now columns, the inner loop is rows
        for (int col = 0; col < grid[0].length; col++) {
            for (int row = 0; row < grid.length; row++) {
                System.out.print(grid[row][col] + " ");
            }
            System.out.println();
        }
    }

    public static void main(String[] args) {
        String[][] myGrid = {
            {"A", "B", "C"},
            {"D", "E", "F"},
            {"G", "H", "I"}
        };
        printArray(myGrid);
    }
}
PrintArray.main(null)
```

## Algorithms That Use 2D Array Traversal

Just like a 1D array, you can run a linear search on a 2D array. It just takes a nested loop instead of a single loop.

```java
public class SearchExample {
    public static boolean search(String[][] chart, String name) {
        for (int r = 0; r < chart.length; r++) {
            for (int c = 0; c < chart[r].length; c++) {
                if (chart[r][c].equals(name)) {
                    return true;
                }
            }
        }
        return false; // Name not found anywhere in the array
    }

    public static void main(String[] args) {
        String[][] chart = {
            {"Alice", "Bob", "Charlie"},
            {"David", "Eve", "Frank"},
            {"Grace", "Hannah", "Isaac"}
        };

        String nameToSearch = "Frank";
        boolean found = search(chart, nameToSearch);

        if (found) {
            System.out.println(nameToSearch + " is in the array.");
        } else {
            System.out.println(nameToSearch + " is not in the array.");
        }
    }
}
SearchExample.main(null)
```

Notice `search` returns the instant it finds a match, from inside the inner loop, rather than finishing the whole traversal. Returning early like this is what makes a linear search efficient: on average, you don't have to look at every element.

## Image Representation with 2D Arrays

A digital image is a grid of small color samples called pixels. Pixels are naturally organized in rows and columns, which is exactly what a 2D array represents. Each element of the array corresponds to one pixel. In a color image, each pixel packs together Red, Green, and Blue (RGB) components; in a grayscale image, each pixel is a single intensity value.

### Traversing and Processing an Image

Traversing an image's 2D array means visiting every pixel with a nested loop: the outer loop over rows, the inner loop over columns. Once you can visit every pixel, you can read or transform its color, which is the basis of almost every image processing operation.

```java
import java.io.*;
import java.net.URL;
import javax.imageio.ImageIO;
import java.awt.image.BufferedImage;

public class ImageProcessing {
    public static void main(String[] args) {
        String imageUrl = "https://example.com/photo.png";
        String outputPath = "images/downloaded.png";

        try {
            URL url = new URL(imageUrl);

            try (InputStream in = url.openStream();
                 OutputStream out = new FileOutputStream(outputPath)) {
                byte[] buffer = new byte[1024];
                int bytesRead;
                while ((bytesRead = in.read(buffer)) != -1) {
                    out.write(buffer, 0, bytesRead);
                }
            }

            BufferedImage image = ImageIO.read(new File(outputPath));
            int width = image.getWidth();
            int height = image.getHeight();
            System.out.println("Image dimensions (width x height): " + width + " x " + height);

            // Process the image using a 2D array: one entry per pixel
            int[][] pixels = new int[width][height];
            for (int x = 0; x < width; x++) {
                for (int y = 0; y < height; y++) {
                    pixels[x][y] = image.getRGB(x, y);
                }
            }
        } catch (IOException e) {
            e.printStackTrace();
        }
    }
}
ImageProcessing.main(null)
```

## Grayscale Conversion with Images

Converting an image to grayscale is a good example of an algorithm that only makes sense once you can traverse a 2D array. For every pixel, you look at its Red, Green, and Blue components and average them into a single intensity value, then store that back into a 2D array of grayscale values.

```java
import java.awt.image.BufferedImage;
import java.io.*;
import javax.imageio.ImageIO;

public class ImageToGrayscaleWithArray {
    public static void main(String[] args) {
        try {
            BufferedImage originalImage = ImageIO.read(new File("images/downloaded.png"));
            int width = originalImage.getWidth();
            int height = originalImage.getHeight();

            // A 2D array to hold one grayscale value per pixel
            int[][] grayscaleArray = new int[width][height];

            for (int x = 0; x < width; x++) {
                for (int y = 0; y < height; y++) {
                    int pixel = originalImage.getRGB(x, y);
                    int red = (pixel >> 16) & 0xFF;
                    int green = (pixel >> 8) & 0xFF;
                    int blue = pixel & 0xFF;
                    int gray = (red + green + blue) / 3;
                    grayscaleArray[x][y] = gray;
                }
            }

            // Build a grayscale image from the array
            BufferedImage grayscaleImage = new BufferedImage(width, height, BufferedImage.TYPE_BYTE_GRAY);
            for (int x = 0; x < width; x++) {
                for (int y = 0; y < height; y++) {
                    int grayValue = grayscaleArray[x][y];
                    int grayPixel = (grayValue << 16) | (grayValue << 8) | grayValue;
                    grayscaleImage.setRGB(x, y, grayPixel);
                }
            }

            ImageIO.write(grayscaleImage, "png", new File("images/grayscale.png"));
            System.out.println("Success! Image converted to grayscale.");
        } catch (IOException e) {
            e.printStackTrace();
        }
    }
}
ImageToGrayscaleWithArray.main(null)
```

Notice this uses two separate traversals over the same dimensions: one to fill `grayscaleArray` from the original pixels, and one to write those values into the new image. That's a common pattern once you're processing real data: read into a structure, then write back out from it.
