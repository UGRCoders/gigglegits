## Standard For Loop

An array in Java is indexed from 0 to `length - 1`. A standard for loop is the most common way to visit every one of those indices, in order, exactly once. That's what makes it a **traversal**: every element gets visited.

**Review on For Loops**
- init: initializes a variable, and runs only once, before the loop starts.
- condition: checked before every iteration; the loop keeps running as long as this is true.
- incr/decr: the increment or decrement statement applied to the variable, runs after every iteration.

## Enhanced For Loop

- The enhanced for loop is also known as the "for each" loop.
- It provides a simplified way to loop through elements in an array, collection, or other iterable data structure.
- You get a copy of each element in order, but you don't get the index, and you can't assign into the loop variable to change what's stored in the array.

The example below fills an array with random numbers, then prints it two different ways: once with a standard for loop, once with an enhanced for loop.

````python
import java.util.Random;

int[] list = new int[6];
Random rand = new Random();

// FOR LOOP: fill the array with random numbers 0-3
for (int i = 0; i < list.length; i++) {
    list[i] = rand.nextInt(4);
}

// ENHANCED FOR LOOP: print every element
for (int element : list) {
    System.out.println(element);
}
````

<div class="callout callout-note">
<h4>Note</h4>

Even though the loops above are for loops, they are NOT traversals, because they skip elements. A traversal has to visit every element in the array exactly once.

## While Loops

A while loop can also traverse an array. Does the loop below visit every element?

````python
int [] list = new int[5];
int index = 0; 

while (index < list.length) 
{
    // Do something
    index ++; 
}
````

<details>
<summary>Show answer</summary>

Yes, this is a traversal. `index` starts at 0 and the loop runs as long as `index < list.length`, which visits every valid index exactly once, just like the standard for loop version.

</details>

This while loop and an equivalent for loop accomplish the same task, but there's one difference: after the loop finishes, the variable `index` from the while loop still exists and is usable, while the variable `i` declared inside a for loop's header does not exist outside the loop. That's because a for loop's init expression declares its variable in the scope of the loop itself, so it goes out of scope the moment the loop ends. A while loop's counter has to be declared before the loop, in the surrounding scope, so it's still around afterward.

## Bounds Errors

When traversing an array, we need to be careful with the indices to avoid an ArrayIndexOutOfBoundsException being thrown. 

<div class="callout callout-hint">
<h4>Attention: Common Mistake</h4>

What is wrong with the for loop and while loop below? Why does each one produce an `ArrayIndexOutOfBoundsException`?

```java
// Version 1: for loop
for (int i = 0; i <= list.length; i++) {
    System.out.println(list[i]);
}

// Version 2: while loop
int index = 0;
while (index <= list.length) {
    System.out.println(list[index]);
    index++;
}
```

</div>

<details>
<summary>Show answer</summary>

Both loops use `<=` instead of `<` as the condition. When `i` (or `index`) equals `list.length`, the loop body still runs, and `list[list.length]` doesn't exist. The last valid index is always `list.length - 1`, so the condition should be `i < list.length`. This off by one mistake is one of the most common bugs in array traversal, both in student code and on the AP exam.

</details>
