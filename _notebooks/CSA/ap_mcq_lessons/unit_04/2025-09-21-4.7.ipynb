# Wrapper Classes

The last lesson used `Integer.parseInt()` to turn a `String` from a file into an `int`. `Integer` is not a primitive type, it's a class, and it exists specifically to make a primitive value behave like an object. This lesson covers why that's necessary, and it directly sets up the next lesson, since ArrayLists can only hold objects.

## Why Primitives Need a Wrapper

Every primitive type has a corresponding wrapper class, a full class that stores the same value but as an object:

| Primitive | Wrapper Class |
| --- | --- |
| `int` | `Integer` |
| `double` | `Double` |
| `char` | `Character` |
| `boolean` | `Boolean` |
| `long` | `Long` |

The reason this matters: Java's generic classes, including `ArrayList<E>`, only work with object types. `ArrayList<int>` is not legal Java, it's a compile error. `ArrayList<Integer>` is. Every `ArrayList<Integer>` you wrote in the last two lessons was only possible because `Integer` is a class, not a primitive.

## Autoboxing and Unboxing

You've actually been using wrapper classes without writing any conversion code yourself. That's because Java automatically converts between a primitive and its wrapper whenever the context calls for it.

```python
import java.util.ArrayList;

public class BoxingExample {
    public static void main(String[] args) {
        ArrayList<Integer> numbers = new ArrayList<>();

        numbers.add(5);          // autoboxing: int 5 becomes Integer.valueOf(5)
        int first = numbers.get(0); // unboxing: Integer becomes int automatically

        System.out.println(first + 1); // 6, ordinary int math works fine after unboxing
    }
}
BoxingExample.main(null);
```

- **Autoboxing** is Java automatically wrapping a primitive value in its wrapper class, `numbers.add(5)` compiles even though `add` expects an `Integer`, because Java converts the `int` for you.
- **Unboxing** is the reverse: converting a wrapper object back to a primitive, so you can use it in ordinary arithmetic.

This happens constantly with ArrayLists of numeric types, and it's why `ArrayList<Integer>` feels like it "just works" like an array of `int`, even though under the hood it's storing full `Integer` objects.

<div class="callout callout-hint">
<h4>Attention: Common Mistake</h4>

Wrapper objects are objects, which means `==` compares references, not values, the same rule as any other object. Run the cell below and see it for yourself:

</div>

```python
Integer a = 200;
Integer b = 200;
System.out.println(a == b);        // may print false: two separate objects
System.out.println(a.equals(b));   // true: compares the actual values
```

Always use `.equals()` to compare wrapper values, or unbox to a primitive first, exactly like the `.equals()` rule you already know for `String` and other objects.

## Converting Between Strings and Numbers

Wrapper classes also provide static methods for converting text to numbers and back, which is exactly what you needed last lesson to turn a line read from a file into a usable number.

```python
public class ConversionExample {
    public static void main(String[] args) {
        String line = "42";
        int value = Integer.parseInt(line);   // String to int
        System.out.println(value + 8);        // 50

        double price = Double.parseDouble("19.99"); // String to double
        System.out.println(price);

        int score = 91;
        String asText = Integer.toString(score); // int to String
        System.out.println("Score: " + asText);
    }
}
ConversionExample.main(null);
```
