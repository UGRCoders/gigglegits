# ArrayList Traversals

This lesson covers how to visit every element of an ArrayList, using a standard for loop, a while loop, and an enhanced for loop, plus the mistakes to watch for when you modify a list while traversing it.

- Traverse an ArrayList with an enhanced for loop
- Safely remove elements from an ArrayList while traversing it
- Recognize common ArrayList traversal mistakes: wrong element type, off by one indexing, and modifying a list mid-traversal

The last lesson covered ArrayList's core methods: `size()`, `add()`, `get()`, `set()`, and `remove()`. Traversing an ArrayList means using those methods together in a loop to visit every element, the same idea as an array traversal, but with `size()` instead of `.length` and `get(i)` instead of `[i]`.

## Traversing with a Standard For Loop

The indices of an ArrayList start at 0, just like an array. If you use an index below 0 or at/above `size()`, you'll get an `IndexOutOfBoundsException`.

```java
import java.util.ArrayList;

public class Main {
    public static void main(String[] args) {
        ArrayList<String> roster = new ArrayList<>();
        roster.add("Hello");
        roster.add("World");
        roster.add("Java");

        int sum = 0;
        for (int i = 0; i < roster.size(); i++) {
            String element = roster.get(i);
            sum += element.length();
        }
        System.out.println(sum); // 5 + 5 + 4 = 14
    }
}
```

This looks almost identical to an array traversal: `roster.size()` plays the role `list.length` played, and `roster.get(i)` plays the role `list[i]` played. The loop still has the same three parts: init, boolean condition, and update, and you can modify any of them to change where the traversal starts, stops, or how it steps.

<div class="callout callout-hint">
<h4>Attention: Common Mistake</h4>

If you try to solve this the "obvious" way, incrementing `i` on every iteration no matter what, you'll skip an element. Removing index 0 shifts everything left by one, so the element that used to be at index 1 is now at index 0, but the loop is about to move on to index 1. You never look at the value that shifted down.

The fix: don't increment `i` when you remove an element, since the next element has slid into the current spot. Only increment when you don't remove.

</div>

<details>
<summary>Show sample answer</summary>

```java
for (int i = 0; i < grades.size(); i++) {
    if (grades.get(i) < 70) {
        grades.remove(i);
        i--; // stay on this index, since the next element just shifted into it
    }
}
```

</details>

## Traversing with an Enhanced For Loop

An enhanced for loop is easier to read and write when you don't need the index, and it avoids off by one mistakes entirely since there's no index to get wrong.

```java
import java.util.ArrayList;
import java.util.List;

public class Main {
    public static void main(String[] args) {
        List<String> roster = new ArrayList<>();
        roster.add("Hello");
        roster.add("World");
        roster.add("Java");

        for (String element : roster) {
            System.out.println(element);
        }
    }
}
```

Note that the enhanced for loop above can't be used to solve the grades problem: since you can't remove elements or track an index inside it, using `remove()` during an enhanced for loop traversal throws a `ConcurrentModificationException`. When you need to modify a list while traversing it, use a standard for loop (or count backward from the end, so removals never shift unvisited elements).

### Common Mistakes

- **Wrong element type.** `ArrayList<Integer>` only accepts `Integer` values (autoboxed from `int`). Trying to `add` a `double` or a `String` is a compile error.
- **Off by one indexing.** Using `<=` instead of `<` against `size()`, or forgetting that `size()` is one more than the last valid index, throws an `IndexOutOfBoundsException`. Same rule as arrays.
- **Modifying the list mid-traversal.** Adding or removing elements while an enhanced for loop is running throws a `ConcurrentModificationException`. If you need to modify while traversing, use an indexed loop.
- **Not checking for null.** If a list might contain `null`, calling a method on an element without checking first throws a `NullPointerException`.
