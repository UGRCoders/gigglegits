## ArrayList Intro

- ArrayLists are dynamic: their size can grow or shrink as needed. Arrays are static: their size is fixed the moment they're created.
- Instead of creating a new, bigger array and copying every element over when you run out of room, an ArrayList handles resizing for you.

| Arrays | ArrayLists |
| --- | --- |
| Fixed length | Resizable length |
| Fundamental Java feature | Part of the `java.util` framework |
| No methods, just `.length` | Class with many built-in methods |
| Can store primitive data | Cannot store primitives directly (needs wrapper classes, see 4.7) |
| Slightly faster | Slightly slower than arrays |
| No import needed | Needs `import java.util.ArrayList;` |

```java
import java.util.ArrayList;

// Declare and initialize an ArrayList of integers
ArrayList<Integer> numbersList = new ArrayList<>();
```

ArrayList objects are created like any other object, except you specify the type of element the list will hold using angle brackets `<>`. In the general form below, `E` stands for whatever element type you choose:

```java
ArrayList<E> list = new ArrayList<E>();
```

You can technically declare an ArrayList without specifying `E`, but you shouldn't. Specifying the type lets the compiler catch type mismatches before your code ever runs, instead of crashing later.

```java
ArrayList list = new ArrayList(); // legal, but avoid this
```

## ArrayList Methods

Iteration statements let you access every element stored in an `ArrayList`; this is called traversing the ArrayList, and it's covered in depth in the next lesson. First, here are the core methods every ArrayList supports. These are part of the AP CSA Java Quick Reference, so it's worth knowing them cold:

| Method | What it does |
| --- | --- |
| `int size()` | Returns the number of elements in the list. |
| `boolean add(E obj)` | Appends `obj` to the end of the list, returns true. |
| `void add(int index, E obj)` | Inserts `obj` at `index`, shifting everything at and after that position one spot to the right. |
| `E get(int index)` | Returns the element at `index`. |
| `E set(int index, E obj)` | Replaces the element at `index` with `obj`, returns the old element. |
| `E remove(int index)` | Removes the element at `index`, shifting everything after it one spot to the left, returns the removed element. |

`ArrayList<E>` is a generic class: whatever type you plug in for `E` is what `get`, `set`, and `add` will accept and return. Using `ArrayList<E>` instead of a raw `ArrayList` lets the compiler catch type mistakes at compile time instead of at runtime.

Run through each method below one at a time:

````python
import java.util.ArrayList;

// --- Size ---
ArrayList<Integer> a1 = new ArrayList<>();
System.out.println(a1.size()); // 0, a new ArrayList starts empty

// --- Adding Items ---
ArrayList<Double> a2 = new ArrayList<>();
a2.add(1.0);
a2.add(2.0);
a2.add(3.0);      // a2 is now [1.0, 2.0, 3.0]
a2.add(1, 4.0);   // insert 4.0 at index 1, shifting the rest right
System.out.println(a2); // [1.0, 4.0, 2.0, 3.0]

// --- Removing Items ---
ArrayList<String> g = new ArrayList<>();
g.add("Hello");
g.add("Hola");
g.add("HeLLO");
g.add("Bonjour");

g.remove(3);              // removes "Bonjour"
String former = g.remove(0); // removes and returns "Hello"
System.out.println(former);  // Hello
System.out.println(g);       // [Hola, HeLLO]

// --- Updating and Accessing Items ---
String helloFormer = g.set(0, "Ciao"); // replaces index 0, returns the old value
System.out.println(helloFormer); // Hola
System.out.println(g);           // [Ciao, HeLLO]

String hello = g.get(1);
System.out.println(hello); // HeLLO
````

### Passing an ArrayList as a Parameter

The one time it's acceptable to use a raw `ArrayList` instead of `ArrayList<E>` is as a method parameter, and only if the method body just calls `.size()` and `.get(index)` without caring about the specific type:

```java
private void accessOnly(ArrayList arr) {
    if (arr.size() > 0) {
        System.out.println(arr.get(0));
    } else {
        System.out.println("Array is empty");
    }
}

ArrayList<Integer> myList = new ArrayList<Integer>();
accessOnly(myList);
```

### Returning an ArrayList from a Method

The return type has to match the ArrayList's declared type:

```java
private ArrayList<String> returnTheSame() {
    ArrayList<String> arr = new ArrayList<String>();
    arr.add("Hello");
    return arr;
}

ArrayList<String> result = returnTheSame();
System.out.println(result); // [Hello]
```
