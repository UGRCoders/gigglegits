## Scenario

You are programming a game where two objects of your choosing (e.g., Robots, Dinosaurs, People) battle. Each object has health and power.

<div class="callout callout-homework" markdown="1">

   **Instance variables:**

   - `int health`
   - `MAKE YOUR OWN`

   **Static variables:**
   - `double fightDuration`

   **Instance methods:**
   - `void attack()`
   - `void printStatus()`
   - `MAKE YOUR OWN`

   **Class methods:**
   - `static int strongerFighter()`
   - `static void beginBattle()`
   - `MAKE YOUR OWN`

2. **In `main`:**
   - Create two objects.
   - Use instance methods to attack and print status.
   - Use static methods to compare, print a fact, and start the battle.

</div>

**Not sure where to start? Build it in this order:**

1. Write the class header and the four instance variables (name, power, health, and your own). Add a constructor that sets them.
2. Add `void attack(YourClass target)` that lowers `target.health` by `this.power`. Test it before adding anything else.
3. Add `void printStatus()` that prints the name, power, and health.
4. Add the `static double fightDuration` variable, then `static int strongerFighter(YourClass a, YourClass b)` that compares two fighters' power and returns which one is stronger.
5. Add `static void beginBattle()` last, once the pieces above already work on their own.

**Quick Check before you code:** why is `fightDuration` static but `health` is not?

<details>
<summary>Show answer</summary>

`health` belongs to one specific fighter, every object needs its own copy. `fightDuration` describes the battle itself, one shared value that every fighter and the `main` method can all read.
