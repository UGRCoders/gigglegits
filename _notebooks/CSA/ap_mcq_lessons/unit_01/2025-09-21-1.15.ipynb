### Java Objects

In Java, objects are essentially instances of classes. They represent real-world entities or concepts and contain both data and behaviors, facilitating organized and modular software design.

### Common Types of Java Classes

In Java, classes are the fundamental building blocks of Object-Oriented Programming (OOP). They define the structure and behavior of objects, serving as templates or blueprints from which objects are created. Java programs typically consist of multiple classes, each designed to fulfill a specific role.

One kind you already know: **utility classes**. A utility class contains only static methods and isn't meant to be instantiated. You've already used one, `Math`, and you're about to use another: the `Arrays` class below has static helper methods like `sort()` that work on arrays without you ever writing `new Arrays()`.

```java
import java.util.Arrays;

public class ArraySortExample {
    public static void main(String[] args) {
        int[] numbers = {5, 2, 9, 1, 5, 6};
        Arrays.sort(numbers);
        System.out.println(Arrays.toString(numbers));
    }
}
String[] words = {"main"};
ArraySortExample.main(words);
```

#### Java Objects

Now that you've seen a utility class in action, step back to how ordinary classes and objects work, since that's the design most of your own code will use. An object in Java is a basic unit of Object-Oriented Programming and represents a real-life entity: `Objects are the instances of a class that are created to use the attributes and methods of a class.` A typical Java program creates many objects, which interact by invoking methods on each other. Every object consists of:

1. **State**: represented by the object's attributes (instance variables).
2. **Behavior**: represented by the object's methods.

### Declaring Objects

When an object of a class is created, the class is said to be **instantiated**. All instances share the attributes and behavior defined by the class, but the values of those attributes (the object's state) are unique to each instance. A single class can have any number of instances.

Declaring a variable, `type name;`, tells the compiler that `name` will refer to data of that `type`. For a primitive variable, this also reserves the right amount of memory. For a reference variable, `type` must be a concrete class name: you can't declare a variable of an abstract class or interface type directly, since neither can be instantiated.

### Initializing a Java Object

The `new` operator instantiates a class: it allocates memory for a new object, invokes the class's constructor, and returns a reference to that memory.

```java
public class Dog {
    public String Name; 
    public int Age;

    public void Bark(int times) {
        for (int i = 0; i < times; i++) {
            System.out.println("Bark"); 
        }
    }

    public static void main(String[] args) { 
        Dog dog = new Dog();

        dog.Bark(5);
    }
}
Dog.main(null);
```

### Calling a Non-Void Method

A non-void method returns a value when you call it, and that value's type is declared in the method's signature (the `String` in `public String getName()` below, for instance).

### String Objects

With the general object model in place, look at a specific built-in object you already use constantly: `String`. Even though you can write string literals directly (`"hello"`), `String` is a real class with its own properties:

- String objects can also be constructed with the `new` keyword: `new String("hello")`.
- Methods that act on a `String` never change the string itself. Strings are **immutable**.
- Strings can be concatenated with the `+` operator, which creates a brand new string rather than modifying either original.
- Escape sequences use a backslash (`\`) to print characters that would otherwise be hard to type: `\"` for a double quote, `\\` for a literal backslash, and `\n` for a newline.

```java
public class Strings_demo {
    public static void main(String[] args) { //main code
        String s1 = "hello"; // the way you're probably used to writing strings
        String s2 = new String("hello"); // an alternate way to create the same string
        System.out.println(s1);// these two println statements print the same thing
        System.out.println(s2);

        String s3 = s1 + s2; // s3 is a concatenation of s1 and s2
        System.out.println(s3);
        System.out.println(s1); // concatenation doesn't change the original String

        // Escape sequences: \" for a quote, \\ for a backslash, \n for a newline
        String s4 = "She said \"hello\" and then typed a backslash: \\\nAll on one line, until the \\n.";
        System.out.println(s4);
    }
}
Strings_demo.main(null);
```

Here are some of the String methods we use:

|Syntax|Definition|
|-|-|
|String(String str)|Constructs a new String object with the same sequence of characters as str|
|int length()|Returns the number of characters in the String|
|String substring(int from, int to)|Returns the substring beginning at from to one before to|
|String substring(int from)|Returns the substring beginning at from to the end|
|int indexOf(String str)|Returns the first occurrence of str, returns -1 if not found|
|boolean equals(String other)|Returns true if this is equal to other and vice versa|
|int compareTo(String other)|Returns a value less than 0 if this is less than other, 0 if they are equal, and a value greater than 0 if this is greater than other|

```java
public class Strings_demo_2 {
    public static void main(String[] args) { //main code
        String s1 = "hello"; // Strings for demonstration
        String s2 = new String("hihihellothere"); // Construction using the method
        System.out.println(s2.length());

        String s3 = s2.substring(4,9); // creating a substring of s1 from index 4 to 9-1
        System.out.println(s3);
        System.out.println(s1); // these should be the same

        System.out.println(s2.indexOf(s1)); //returns the first occurence of hello in hihihellothere (4)

        System.out.println(s2.equals(s1)); // returns if hello is the same as hihihellothere (it's not)
        System.out.println(s2.compareTo(s1)); // compares hihihellothere to hello, since it greater since h=h and i>e, this returns a positive number
        System.out.println(s1.compareTo(s2)); // the reversal of the above, returns a negative number
        System.out.println(s3.equals(s1)); // returns if hello is the same as hello (it's)
        System.out.println(s3.compareTo(s1)); // compares hello to hello, since they are the same, this returns a 0

        System.out.println(s1);// none of these methods affect the original strings, since Strings are immutatble
        System.out.println(s2);
        System.out.println(s3);
    }
}
Strings_demo_2.main(null);
```

### Wrapper Classes

You've just seen `String`, a reference type built into Java. This section covers two more reference types you'll use constantly: `Integer` and `Double`, which wrap the `int` and `double` primitives you've used since Unit 1 began.

- Integer (not int: notice the non-shortened word and capitalization)
    - int is a primitive data type, and Integer is a wrapper class

|Syntax|Definition|
|-|-|
|Integer(int value)|Constructs a new Integer object representing the value|
|Integer.MIN_VALUE|the minimum value represented by a int or Integer|
|Integer.MAX_VALUE|the maximum value represented by a int or Integer|
|int intValue()|Returns the value of Integer as int|

- Double (not double)
    - double is a primitive data type, Double is a wrapper class

|Syntax|Definition|
|-|-|
|Double(double value)|Constructs a new Double object representing the value|
|double doubleValue()|Returns the value of the Double as a double|

#### Why Use Wrapper Classes?

Wrapper classes let you use primitive values (like `int` and `boolean`) anywhere Java expects an object, since primitives themselves aren't objects and some parts of the language, like collections, only work with objects.

#### Autoboxing and Unboxing

As of Java 5.0, autoboxing and unboxing are built into the language.

**Autoboxing** is the automatic conversion the compiler performs from a primitive value to its wrapper class.

**Unboxing** is the reverse: converting a wrapper object back to its primitive value.

```java
public class Unboxing_stream {
    public static void main(String[] args) { //main code
        int i1 = 10; // primitive data types
        double d1 = 25.5;

        Integer I1 = new Integer(i1); // explicit construction using the constructor
        Double D1 = new Double(d1);

        System.out.println(i1); // The values correspond to their original values
        System.out.println(I1);
        System.out.println(d1);
        System.out.println(D1);

        int i2 = I1.intValue()+5; //usage of the intValue function
        System.out.println(i2);
        double d2 = D1.doubleValue()-1.3; //usage of the doubleValue function
        System.out.println(d2);

        System.out.println(Integer.MIN_VALUE); // the actual values of the MIN_VALUE and MAX_VALUE
        System.out.println(Integer.MAX_VALUE);

        Integer I2 = i1; // autoboxing bypasses the need for constructor methods
        Double D2 = d1;
        System.out.println(I2);
        System.out.println(D2);

        int i3 = I2+10; // unboxing without using any methods
        double d3 = D2-1.3;
        System.out.println(i3);
        System.out.println(d3);
    }
}
Unboxing_stream.main(null);
```

### The Math Class

You met the `Math` class back in 1.11; here's a quick review of the methods most likely to show up mixed in with everything else this unit covered.
- Math methods are static, they belong to the class instead of an object and can be used without creating an object
- The Math class is already imported as part of java.lang.Math
    - However, you can use the methods without the Math prefix with the import command `import static java.lang.Math.*`

|Syntax|Definition|
|-|-|
|static int abs(int x)|Returns the absolute value of a int|
|static double abs(double x)|Returns the absolute value of a double|
|static double pow(double base, double exponent)|Returns the value of base to the exponent power|
|static double random()|Returns a random double between 0.0 and 1.0|

```java
import static java.lang.Math.*; // implicit importation 

public class maths_demo {
    public static void main(String[] args) { //main code
        double a=12.5; // ints we will use
        int b=-5;
        int c=4;

        System.out.println(abs(a)); // prints the absolute values of a and b
        System.out.println(abs(b)); // should have no effect on a but b should be turned to its opposite (the positive number)

        System.out.println(pow(b,c)); // prints the result of b to c

        System.out.println(random()*10>4); // returns true if a random number from 0 to 10 is greater than 4, this should return different values with different runs.
    }
}

maths_demo.main(null);
```
